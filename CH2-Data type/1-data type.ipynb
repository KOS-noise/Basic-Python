Data type
<br>

Python Programs handle vast amounts of data. The data have been categorized based on their functions and roles to make them easily accessible for developers.
<br>

So, these types are called data type(자료형).
Most fundamental data types have String(문자열), 숫자(number), 불(boolean).

Type
<br>

Data type is about a form of data. In python, we use function 'type()' to confirm a form of data. Just like function 'print()', there are parentheses following it. If we put data in parentheses, We can confirm what data type the data have.

```python
print(type("안녕하세요"))
```

```python
print(type(273))
```

Make String
<br>

In Programming language, what is called 'String' which word sequenced like this.<br>

Anything enclosed in quotation(따옴표) marks is defined as a string.

String is enclosed in double quotation(큰 따옴표) marks. 

And also, we can use single quotation marks(작은 따옴표) too.

Including quotation marks within a string
<br>

For example, let's assume you want to create a string that includes double quotation marks, like this

```python
Please call me "KOS noise"
```

As I mentioned before, We can make this in Python by using double quotation marks.

```python
print("Please call me "KOS noise"")
```

What were the result? 
<br>
Oops. Error occurred like this.

SyntaxError: invalid syntax

Python programming language recognize like this.
<br>

"Please call me " + KOS noise + ""

So, what should we do if we want to include double quotation marks inside a string?

It's simple. We just have to use single quotation.

```python
print('Please call me "KOS noise"')
```

Conversely, what if we want to include a single quote inside a string?

Likewise, we can use double quotation like this.

```python
print("Please call me 'KOS noise'")
```

<h1 style="color:#0969da; font-size:2.4em; font-weight:700; letter-spacing:-0.02em; margin-bottom:0.35em;">Make String by using Escape character</h1>

<p style="line-height:1.8; font-size:1.05em; color:#1f2328;">
Regardless of how we create a string, there is a way to easily include double or single quotes whenever we need them.
</p>

<p style="line-height:1.8; color:#1f2328;">
That method called escape character. It means a special string which is a combination of backslash(역슬래시 \).
</p>

<p style="line-height:1.8; color:#1f2328;">
Using by this, Python recognize just quotation. Not a mark that make a string.
</p>

<ul style="line-height:1.9; color:#1f2328;">
<li><code>\"</code> : means double quotation mark (큰따옴표를 의미한다)</li>
<li><code>\'</code> : means single quotation mark (작은 따옴표를 의미한다)</li>
</ul>

```python
print("Please call me \"KOS noise\"")
```

```python
print('He said \'Im hungry\'')
```

<h2 style="color:#8250df; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">In addition to this, there are various escape character.</h2>

<ul style="line-height:1.9; color:#1f2328;">
<li><code>\n</code> : means a line break (줄바꿈)</li>
<li><code>\t</code> : means tab (탭)</li>
<li><code>\\</code> : means backslash(\)</li>
</ul>

```python
print("Hello\nWorld")
```

```python
print("Hello\tWorld")
```

```python
print("Hello\\World")
```

<h2 style="color:#1a7f37; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">And also, python supports a feature called multi-line strings.</h2>

<p style="line-height:1.8; color:#1f2328;">
This uses a symbol consisting of three repeated quotation marks.
</p>

```python
print(""" 동해물과 백두산이 마르고 닳도록
하느님이 보우하사 우리나라 만세
무궁화 삼천리 화려강산 대한사람
대한으로 길이 보전하세 """)
```

<h1 style="color:#0969da; font-size:2.4em; font-weight:700; letter-spacing:-0.02em; margin-bottom:0.35em;">String Operator</h1>

<p style="line-height:1.8; font-size:1.05em; color:#1f2328;">
We can operate strings with operators.<br>
The most basic ones are <code>+</code> and <code>*</code>.
</p>

<h2 style="color:#8250df; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">String Concatenation Operator <code>+</code></h2>

<p style="line-height:1.8; color:#1f2328;">
The <code>+</code> operator connects two strings into one string.
</p>

```python
print("Hello" + "World")
```

```python
print("Hello" + " " + "Python")
```

<h2 style="color:#1a7f37; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">String Repetition Operator <code>*</code></h2>

<p style="line-height:1.8; color:#1f2328;">
The <code>*</code> operator repeats a string as many times as the number on the right.
</p>

```python
print("Hello" * 3)
```

```python
print("=" * 10)
```

<h2 style="color:#0550ae; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Character Selection Operator &amp; Index</h2>

<p style="line-height:1.8; color:#1f2328;">
The <strong>character selection operator</strong> selects a single character from inside a string.<br>
You put the position of the character you want inside square brackets <code>[]</code>, and that number is called an <strong>index</strong>.
</p>

<p style="line-height:1.8; color:#1f2328; margin-top:1em;">
Programming languages generally use one of two index styles:
</p>

- <strong>Zero-based indexing</strong> — counting starts at <code>0</code>
- <strong>One-based indexing</strong> — counting starts at <code>1</code>

<blockquote>
<p>Python uses <strong>zero-based indexing</strong>.</p>
</blockquote>

<h3 style="color:#0969da; font-size:1.25em; margin-top:1.2em;">Example: indexing <code>"Hello"</code></h3>

| Character | H | e | l | l | o |
|-----------|---|---|---|---|---|
| Index     | 0 | 1 | 2 | 3 | 4 |

```python
print("Hello"[0])
print("Hello"[1])
print("Hello"[4])
```

<h2 style="color:#8250df; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Reverse Index (Negative Index)</h2>

<p style="line-height:1.8; color:#1f2328;">
Python also supports <strong>reverse indexing</strong>.<br>
Counting starts from the last character with <code>-1</code>, then <code>-2</code>, <code>-3</code>, and so on.
</p>

<h3 style="color:#8250df; font-size:1.25em; margin-top:1.2em;">Example: reverse indexing <code>"Hello"</code></h3>

| Character     | H  | e  | l  | l  | o  |
|---------------|----|----|----|----|----|
| Index         | 0  | 1  | 2  | 3  | 4  |
| Reverse Index | -5 | -4 | -3 | -2 | -1 |

```python
print("Hello"[-1])
print("Hello"[-2])
print("Hello"[-5])
```

<h2 style="color:#1a7f37; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Indexing vs Slicing</h2>

| Term | Meaning |
|------|---------|
| **Indexing** | Referring to a character at a specific position in a string |
| **Slicing** | Extracting a part of a string with the <code>[:]</code> operator |

<blockquote>
<p style="line-height:1.8;">
문자열의 특정 위치에 있는 문자를 참조하는 것을 <strong>인덱싱(indexing)</strong>이라 하고,<br>
<code>[:]</code> 연산자를 이용해 문자열의 일부를 추출하는 것을 <strong>슬라이싱(slicing)</strong>이라 한다.
</p>
</blockquote>

<h2 style="color:#0969da; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">String Range Selection Operator (Slicing)</h2>

<p style="line-height:1.8; color:#1f2328;">
Slicing uses <code>[start:end]</code>.<br>
It starts at <code>start</code>, and stops <strong>before</strong> <code>end</code>. (<code>end</code> is not included.)
</p>

| Syntax | Meaning |
|--------|---------|
| <code>[0:2]</code> | from index 0 to index 1 |
| <code>[1:]</code> | from index 1 to the end |
| <code>[:3]</code> | from the beginning to index 2 |
| <code>[:]</code> | the whole string |

```python
print("안녕하세요"[0:2])
```

```python
print("안녕하세요"[1:])
```

```python
print("안녕하세요"[:3])
```

```python
print("안녕하세요"[:])
```

<h2 style="color:#d1242f; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">IndexError</h2>

<p style="line-height:1.8; color:#1f2328;">
If you select an index that does not exist in the string, Python raises an <strong>IndexError</strong> exception.
</p>

<blockquote>
<p style="line-height:1.8;">
<code>"안녕하세요"</code> has only 5 characters, so the valid indexes are <code>0</code> to <code>4</code> (or <code>-1</code> to <code>-5</code>).<br>
Index <code>10</code> is out of range.
</p>
</blockquote>

```python
print("안녕하세요"[10])
```

<h2 style="color:#0969da;">How to get the length of string</h2>

We can get the length of a string with the function `len()`.

Put a string inside the parentheses. Then Python returns how many characters it has.

> `len()` counts characters. It does not count from 0 like an index.

```python
print(len("안녕하세요"))
print(len("Hello"))
```

<h1 style="color:#0969da; font-size:2.4em; font-weight:700; letter-spacing:-0.02em; margin-bottom:0.35em;">Number</h1>

<p style="line-height:1.8;">Python has two basic number types: integer and floating-point number.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">파이썬의 기본 숫자 자료형은 정수와 실수 두 가지입니다.</p>

<p style="line-height:1.8;">An integer(정수) is a number without a decimal point, like <code>273</code> or <code>-52</code>.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">정수(integer)는 소수점이 없는 숫자입니다. 예를 들어 <code>273</code>이나 <code>-52</code>입니다.</p>

<p style="line-height:1.8;">A floating-point number(실수) is a number with a decimal point, like <code>52.273</code> or <code>0.5</code>.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">실수(floating-point)는 소수점이 있는 숫자입니다. 예를 들어 <code>52.273</code>이나 <code>0.5</code>입니다.</p>

<h2 style="color:#8250df; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Check the type with <code>type()</code></h2>

<p style="line-height:1.8;">We can use the function <code>type()</code> to check whether a number is an integer or a floating-point number.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">함수 <code>type()</code>을 사용하면 숫자가 정수인지 실수인지 확인할 수 있습니다.</p>

<p style="line-height:1.8;">If we put a number inside the parentheses, Python shows the data type.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">괄호 안에 숫자를 넣으면, 파이썬이 그 자료형을 보여 줍니다.</p>

```python
print(type(52))
print(type(52.273))
```

<p style="line-height:1.8;"><code>52</code> is <code>int</code>. That means integer.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;"><code>52</code>는 <code>int</code>입니다. 정수를 의미합니다.</p>

<p style="line-height:1.8;"><code>52.273</code> is <code>float</code>. That means floating-point number.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;"><code>52.273</code>는 <code>float</code>입니다. 실수를 의미합니다.</p>

<h2 style="color:#1a7f37; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Number Operator</h2>

<p style="line-height:1.8;">We can calculate numbers with operators.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">연산자를 사용해서 숫자를 계산할 수 있습니다.</p>

<p style="line-height:1.8;">The four basic operators are addition <code>+</code>, subtraction <code>-</code>, multiplication <code>*</code>, and division <code>/</code>.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">기본 연산자 네 가지는 덧셈 <code>+</code>, 뺄셈 <code>-</code>, 곱셈 <code>*</code>, 나눗셈 <code>/</code>입니다.</p>

```python
print("5 + 7 =", 5 + 7)
print("5 - 7 =", 5 - 7)
print("5 * 7 =", 5 * 7)
print("5 / 7 =", 5 / 7)
```

<p style="line-height:1.8;">In Python 3, the division operator <code>/</code> always returns a floating-point number.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">파이썬 3에서 나눗셈 연산자 <code>/</code>는 항상 실수를 반환합니다.</p>

<p style="line-height:1.8;">For example, <code>5 / 7</code> is not an integer. It is a float.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">예를 들어 <code>5 / 7</code>은 정수가 아니라 실수입니다.</p>

<h2 style="color:#0550ae; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Integer Division Operator <code>//</code></h2>

<p style="line-height:1.8;">The <code>//</code> operator divides two numbers and keeps only the integer part.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;"><code>//</code> 연산자는 두 숫자를 나눈 뒤 정수 부분만 남깁니다.</p>

<p style="line-height:1.8;">The digits after the decimal point are thrown away.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">소수점 아래는 버립니다.</p>

```python
print("3 / 2 =", 3 / 2)
print("3 // 2 =", 3 // 2)
```

<p style="line-height:1.8;"><code>3 / 2</code> is <code>1.5</code>.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;"><code>3 / 2</code>는 <code>1.5</code>입니다.</p>

<p style="line-height:1.8;"><code>3 // 2</code> is <code>1</code>, because only the integer part remains.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;"><code>3 // 2</code>는 정수 부분만 남으므로 <code>1</code>입니다.</p>

<h2 style="color:#8250df; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Remainder Operator <code>%</code></h2>

<p style="line-height:1.8;">The <code>%</code> operator returns the remainder after division.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;"><code>%</code> 연산자는 나눗셈을 한 뒤 나머지를 반환합니다.</p>

<p style="line-height:1.8;">For example, <code>5 % 2</code> is <code>1</code>, because 5 divided by 2 is 2 with remainder 1.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">예를 들어 <code>5 % 2</code>는 <code>1</code>입니다. 5를 2로 나누면 몫이 2이고 나머지가 1이기 때문입니다.</p>

```python
print("5 % 2 =", 5 % 2)
print("6 % 2 =", 6 % 2)
```

<h2 style="color:#1a7f37; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Exponent Operator <code>**</code></h2>

<p style="line-height:1.8;">The <code>**</code> operator means power.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;"><code>**</code> 연산자는 제곱을 의미합니다.</p>

<p style="line-height:1.8;"><code>2 ** 2</code> means 2 to the power of 2, so the result is <code>4</code>.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;"><code>2 ** 2</code>는 2의 2제곱이므로 결과는 <code>4</code>입니다.</p>

<p style="line-height:1.8;"><code>2 ** 3</code> means 2 to the power of 3, so the result is <code>8</code>.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;"><code>2 ** 3</code>는 2의 3제곱이므로 결과는 <code>8</code>입니다.</p>

```python
print("2 ** 2 =", 2 ** 2)
print("2 ** 3 =", 2 ** 3)
```

<h2 style="color:#0969da; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Operator Priority</h2>

<p style="line-height:1.8;">Multiplication and division are calculated before addition and subtraction.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">곱셈과 나눗셈은 덧셈과 뺄셈보다 먼저 계산됩니다.</p>

<p style="line-height:1.8;">If we want to change the order, we use parentheses <code>()</code>.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">순서를 바꾸고 싶으면 괄호 <code>()</code>를 사용합니다.</p>

```python
print("5 + 3 * 2 =", 5 + 3 * 2)
print("(5 + 3) * 2 =", (5 + 3) * 2)
```

<p style="line-height:1.8;"><code>5 + 3 * 2</code> is <code>11</code>, because <code>3 * 2</code> is calculated first.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;"><code>5 + 3 * 2</code>는 <code>3 * 2</code>를 먼저 계산하므로 <code>11</code>입니다.</p>

<p style="line-height:1.8;"><code>(5 + 3) * 2</code> is <code>16</code>, because the parentheses are calculated first.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;"><code>(5 + 3) * 2</code>는 괄호를 먼저 계산하므로 <code>16</code>입니다.</p>

<h2 style="color:#d1242f; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">TypeError: string + number</h2>

<p style="line-height:1.8;">The <code>+</code> operator works with strings, and it also works with numbers.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;"><code>+</code> 연산자는 문자열에도 쓰이고, 숫자에도 쓰입니다.</p>

<p style="line-height:1.8;">But we cannot mix a string and a number with <code>+</code>.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">하지만 문자열과 숫자를 <code>+</code>로 섞어서 쓸 수는 없습니다.</p>

<p style="line-height:1.8;">If we try <code>"안녕" + 5</code>, Python raises a <strong>TypeError</strong>.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;"><code>"안녕" + 5</code>를 하면 파이썬은 <strong>TypeError</strong>를 발생시킵니다.</p>

```python
print("안녕" + "하세요")
print(5 + 7)
```

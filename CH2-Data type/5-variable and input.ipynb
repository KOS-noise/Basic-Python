<h1 style="color:#0969da; font-size:2.4em; font-weight:700; letter-spacing:-0.02em; margin-bottom:0.35em;">Make Variable</h1>

<p style="line-height:1.8;">A variable is a name that stores a value.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">변수는 값을 저장하는 이름입니다.</p>

````python
pi = 3.14159265
pi
````

<h2 style="color:#8250df; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Three ways to use a variable</h2>

<p style="line-height:1.8;">There are three ways to use a variable.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">변수를 사용하는 방법은 세 가지입니다.</p>

<p style="line-height:1.8;"><strong>First, declare a variable.</strong></p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">첫째, 변수를 선언합니다.</p>

<p style="line-height:1.8;"><strong>Second, assign a value to the variable.</strong></p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">둘째, 변수에 값을 대입합니다.</p>

<p style="line-height:1.8;">We declared the variable name <code>pi</code>.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">변수 이름을 <code>pi</code>로 선언했습니다.</p>

<p style="line-height:1.8;">So we put <code>3.14159265</code> in <code>pi</code>.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">그래서 <code>pi</code>에 <code>3.14159265</code>를 넣습니다.</p>

<p style="line-height:1.8;">Here, the mark <code>=</code> does not mean “the same”. It means “assign the right value to the left”.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">여기서 <code>=</code>는 ‘같다’가 아닙니다. 오른쪽 값을 왼쪽에 대입한다는 뜻입니다.</p>

<p style="line-height:1.8;"><strong>Third, reference the variable.</strong></p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">셋째, 변수를 참조합니다.</p>

<p style="line-height:1.8;">It means pull the value stored in the variable. So we call this <strong>variable reference</strong>.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">변수 안에 들어 있는 값을 꺼내 쓰는 것입니다. 그래서 이것을 <strong>변수 참조</strong>라고 부릅니다.</p>

````python
# 원의 둘레와 넓이 구하기

pi = 3.14159265
r = 10

# reference variable
print("원주율 =", pi)
print("반지름 =", r)
print("원의 둘레 =", 2 * pi * r)
print("원의 넓이 =", pi * r * r)
````

<h2 style="color:#1a7f37; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Plus content</h2>

<p style="line-height:1.8;">In other programming languages like C, C++, Java, and C#, we must declare the data type of a variable before we use it.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">C, C++, Java, C# 같은 다른 프로그래밍 언어에서는 변수를 쓰기 전에 자료형을 반드시 선언해야 합니다.</p>

<p style="line-height:1.8;">Python does not need that. We just write the name and assign a value.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">파이썬은 그럴 필요가 없습니다. 이름만 쓰고 값을 대입하면 됩니다.</p>

<p style="line-height:1.8;">In Java or C, it looks like this:</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">Java나 C에서는 이렇게 씁니다.</p>

```c
int pi;
```

<h2 style="color:#0550ae; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Compound Assignment Operator</h2>

<p style="line-height:1.8;">A compound assignment operator calculates and assigns at the same time.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">복합 대입 연산자는 계산과 대입을 한 번에 합니다.</p>

<p style="line-height:1.8;"><code>a += 10</code> means <code>a = a + 10</code>.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;"><code>a += 10</code>은 <code>a = a + 10</code>과 같습니다.</p>

````python
a = 0
a += 10
print(a)
````

<p style="line-height:1.8;">These are the compound assignment operators:</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">복합 대입 연산자는 다음과 같습니다.</p>

| Operator | Meaning |
|----------|---------|
| <code>+=</code> | add, then assign |
| <code>-=</code> | subtract, then assign |
| <code>*=</code> | multiply, then assign |
| <code>/=</code> | divide, then assign |
| <code>%=</code> | remainder, then assign |
| <code>**=</code> | power, then assign |

<p style="line-height:1.8;">Strings can use this too.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">문자열에도 사용할 수 있습니다.</p>

<p style="line-height:1.8;"><code>+=</code> : concatenate the string, then assign</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;"><code>+=</code> : 문자열을 연결한 뒤 대입합니다.</p>

<p style="line-height:1.8;"><code>*=</code> : repeat the string, then assign</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;"><code>*=</code> : 문자열을 반복한 뒤 대입합니다.</p>

````python
string = "Hello"

string += "?"

print(string)
````

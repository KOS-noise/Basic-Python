<h1 style="color:#0969da; font-size:2.4em; font-weight:700; letter-spacing:-0.02em; margin-bottom:0.35em;">User Input: <code>input()</code></h1>

<p style="line-height:1.8;">Programs can receive the user's input. We use the function <code>input()</code> to receive it.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">프로그램은 사용자의 입력을 받을 수 있습니다. 함수 <code>input()</code>을 사용해서 받습니다.</p>

<p style="line-height:1.8;">The string in the parentheses is called a <strong>prompt string</strong>.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">괄호 안에 넣는 문자열을 <strong>프롬프트 문자열</strong>이라고 합니다.</p>

<p style="line-height:1.8;">This means it asks the user to type something.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">사용자에게 입력을 요청한다는 뜻입니다.</p>

```python
input("How are you today?")
```

<p style="line-height:1.8;">If we run this code, the program shows us <code>How are you today?</code> and waits. It does not exit.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">이 코드를 실행하면 프로그램이 <code>How are you today?</code>를 보여 주고 기다립니다. 종료되지 않습니다.</p>

<p style="line-height:1.8;">When a program pauses like this during execution, it is called <strong>blocking</strong>.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">실행 중에 프로그램이 잠시 멈추는 것을 <strong>블로킹(blocking)</strong>이라고 합니다.</p>

```python
# input_error.py

string = input("Enter a string: ")

print("input + 100:", string + 100)
```

<h2 style="color:#d1242f; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Why did the error occur?</h2>

<p style="line-height:1.8;">Even if we enter a number, the program treats it as a string.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">숫자를 입력해도, 프로그램은 그 숫자를 문자열로 인식합니다.</p>

<p style="line-height:1.8;">In Python, strings must be used with strings, and numbers must be used with numbers.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">파이썬은 문자열은 문자열끼리 연산해야 하고, 숫자는 숫자끼리 연산해야 합니다.</p>

<h2 style="color:#8250df; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Converting a string to a number</h2>

<p style="line-height:1.8;">So we have to convert the string to a number to calculate.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">그래서 숫자를 계산하려면 문자열을 숫자로 바꿔야 합니다.</p>

<p style="line-height:1.8;">This is called <strong>casting</strong>.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">이것을 <strong>캐스트(cast)</strong>라고 합니다.</p>

<p style="line-height:1.8;">Fortunately, Python provides functions for this.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">다행히 파이썬은 이를 해결하는 함수를 제공합니다.</p>

- <code>int()</code> : convert to an integer (정수로 변환)
- <code>float()</code> : convert to a floating-point number (실수로 변환)

```python
string_a = input("Enter a number: ")
int_a = int(string_a)

string_b = input("Enter another number: ")
int_b = int(string_b)

print("string data:", string_a + string_b)
print("integer data:", int_a + int_b)
```

```python
output_a = int("52")
print(type(output_a))
```

```python
# combining function int() and float()

input_a = float(input("First number: "))
input_b = float(input("Second number: "))

print("addition result: ", input_a + input_b)
print("subtraction result: ", input_a - input_b)
print("multiplication result: ", input_a * input_b)
print("division result: ", input_a / input_b)
```

<h2 style="color:#d1242f; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">ValueError</h2>

<p style="line-height:1.8;">If we try to convert a value that cannot be converted, a <strong>ValueError</strong> occurs.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">변환할 수 없는 값을 변환하려고 하면 <strong>ValueError</strong>가 발생합니다.</p>

<p style="line-height:1.8;">There are two cases of this error.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">이 오류는 두 가지 경우가 있습니다.</p>

<p style="line-height:1.8;"><strong>First,</strong> converting a non-numeric value into a number.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;"><strong>첫째,</strong> 숫자가 아닌 값을 숫자로 변환하려고 할 때입니다.</p>

<p style="line-height:1.8;"><strong>Second,</strong> using <code>int()</code> on a string that has a decimal point.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;"><strong>둘째,</strong> 소수점이 있는 문자열을 <code>int()</code>로 변환하려고 할 때입니다.</p>

```python
# first case
int("안녕하세요")
```

```python
# second case
int("52.273")
```

<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">함수 <code>str()</code>을 이렇게 사용하면 됩니다.</p>

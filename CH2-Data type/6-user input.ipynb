User Input : input()

Programs can receive user's input. We have to use the function 'input()' to receive it.

In this, string what is put in parentheses is called 'prompt string'.
This mean require to user to input.

```python
input("How are you today?")
```

If we play this code, program show us 'How are you today?' and it waited in status that program not exit.
Like this, where a program pauses temporarily during execution, is called "blocking."

```python
# input_error.py

string = input("Enter a string: ")

print("input + 100:", string + 100)
```

Converting a string to a number

So, we have to convert string to number to operate numbers.
This called 'cast'. 

Fortunately, Python provide a function to solve this.

- int() 
- float()

```python
string_a = input("Enter a number: ")
int_a = int(string_a)

string_b = input("Enter another number: ")
int_b = int(string_b)

print("string data:", string_a + string_b)
print("integer data:", int_a + int_b)
```

```python
output_a = int("52")
print(type(output_a))
```

```python
# combining function int() and float()

input_a = float(input("First number: "))
input_b = float(input("Second number: "))

print("addition result: ", input_a + input_b)
print("subtraction result: ", input_a - input_b)
print("multiplication result: ", input_a * input_b)
print("division result: ", input_a / input_b)
```

ValueError 예외

When we conversion data type that cannot conversion, ValueError is occured. 
There are two case of this error.

First, When attempting to convert a non-numeric value into a number.

Second, When attempting to convert a string representing a number with a decimal point using the int() function

```python
# first case
int("안녕하세요")
```

```python
# second case
int("52.273")
```

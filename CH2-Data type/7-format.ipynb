<h1 style="color:#0969da; font-size:2.4em; font-weight:700; letter-spacing:-0.02em; margin-bottom:0.35em;">format()</h1>

<p style="line-height:1.8;"><code>format()</code> puts values into a string. <code>{}</code> is the placeholder.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;"><code>format()</code>은 값을 문자열 안에 넣습니다. <code>{}</code>가 그 자리입니다.</p>

```python
print("{}".format(10))
print("{} {}".format(10, 20))
print("{} {} {}".format(10, 20, 30))
```

<p style="line-height:1.8;">You can also choose the order with numbers inside <code>{}</code>.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;"><code>{}</code> 안에 숫자를 넣으면 순서를 정할 수 있습니다.</p>

```python
print("{0} {1}".format(10, 20))
print("{1} {0}".format(10, 20))
```

<h2 style="color:#d1242f; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">IndexError</h2>

<p style="line-height:1.8;">If there are more <code>{}</code> than arguments in <code>format()</code>, Python raises an <strong>IndexError</strong>.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;"><code>{}</code> 개수가 <code>format()</code> 매개변수보다 많으면 <strong>IndexError</strong>가 발생합니다.</p>

```python
print("{} {}".format(1))
```

<h2 style="color:#8250df; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Integer format: <code>{:d}</code></h2>

<p style="line-height:1.8;"><code>d</code> means integer. The number after <code>:</code> is the width.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;"><code>d</code>는 정수입니다. <code>:</code> 뒤 숫자는 칸 수입니다.</p>

<p style="line-height:1.8;"><code>{:05d}</code> fills empty spaces with <code>0</code>.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;"><code>{:05d}</code>는 빈칸을 <code>0</code>으로 채웁니다.</p>

```python
print("{:d}".format(52))
print("{:d}".format(-52))
print("{:10d}".format(52))
print("{:10d}".format(-52))
print("{:05d}".format(52))
print("{:05d}".format(-52))
```

<p style="line-height:1.8;"><code>{:10d}</code> uses 10 spaces, aligned to the right. <code>-52</code> needs one extra space for the minus sign, so it looks a little different.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;"><code>{:10d}</code>는 10칸을 쓰고 오른쪽 정렬입니다. <code>-52</code>는 마이너스 기호가 한 칸을 써서 조금 다르게 보입니다.</p>

<p style="line-height:1.8;"><code>{:05d}</code> with <code>-52</code> becomes <code>-0052</code>. The sign is kept, then zeros fill the rest.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;"><code>-52</code>를 <code>{:05d}</code>로 쓰면 <code>-0052</code>가 됩니다. 기호를 남기고 나머지를 0으로 채웁니다.</p>

<h2 style="color:#1a7f37; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Print the sign</h2>

<p style="line-height:1.8;"><code>+</code> always shows the sign. A space <code> </code> shows a blank for plus, and <code>-</code> for minus.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;"><code>+</code>는 기호를 항상 보여 줍니다. 공백 <code> </code>은 양수일 때 빈칸, 음수일 때 <code>-</code>를 붙입니다.</p>

```python
output_a = "{:+d}".format(52)
output_b = "{:+d}".format(-52)
output_c = "{: d}".format(52)
output_d = "{: d}".format(-52)

print(output_a)
print(output_b)
print(output_c)
print(output_d)
```

<h2 style="color:#0550ae; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">format04.py : combination</h2>

<p style="line-height:1.8;"><code>=</code> puts the sign on the left, then fills spaces or zeros.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;"><code>=</code>는 기호를 왼쪽에 두고, 그 뒤를 공백이나 0으로 채웁니다.</p>

```python
print("{:+5d}".format(52))
print("{:+5d}".format(-52))
print("{:=+5d}".format(52))
print("{:=+5d}".format(-52))
print("{:=+05d}".format(52))
print("{:=+05d}".format(-52))
```

<p style="line-height:1.8;"><code>{:+5d}</code> : 5 spaces, sign next to the number. <code>52</code> → <code>&nbsp;&nbsp;+52</code> / <code>-52</code> → <code>&nbsp;&nbsp;-52</code></p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;"><code>{:+5d}</code> : 5칸, 기호가 숫자 옆에 붙습니다.</p>

<p style="line-height:1.8;"><code>{:=+5d}</code> : sign first, then spaces. <code>52</code> → <code>+&nbsp;&nbsp;52</code> / <code>-52</code> → <code>-&nbsp;&nbsp;52</code></p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;"><code>{:=+5d}</code> : 기호가 맨 앞, 그다음 공백입니다.</p>

<p style="line-height:1.8;"><code>{:=+05d}</code> : sign first, then zeros. Both <code>+0052</code> and <code>-0052</code>.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;"><code>{:=+05d}</code> : 기호 다음에 0을 채웁니다. <code>+0052</code>, <code>-0052</code>.</p>

<h2 style="color:#8250df; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Floating-point format: <code>{:f}</code></h2>

<p style="line-height:1.8;"><code>f</code> means float. Default is 6 digits after the point.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;"><code>f</code>는 실수입니다. 기본은 소수점 아래 6자리입니다.</p>

```python
print("{:f}".format(52.273))
print("{:15f}".format(52.273))
print("{:+15f}".format(52.273))
print("{:+015f}".format(52.273))
```

<h2 style="color:#1a7f37; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Digits after the decimal point</h2>

<p style="line-height:1.8;"><code>{:15.3f}</code> means width 15, and 3 digits after the point.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;"><code>{:15.3f}</code>는 전체 15칸, 소수점 아래 3자리입니다.</p>

```python
print("{:15.3f}".format(52.273))
```

<h2 style="color:#0550ae; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Remove extra zeros: <code>{:g}</code></h2>

<p style="line-height:1.8;"><code>g</code> drops zeros that are not needed.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;"><code>g</code>는 의미 없는 0을 뺍니다.</p>

```python
print("{:g}".format(52.273))
print("{:g}".format(52.270))
```

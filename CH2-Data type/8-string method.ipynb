<h1 style="color:#0969da; font-size:2.4em; font-weight:700; letter-spacing:-0.02em; margin-bottom:0.35em;">String Method</h1>

<p style="line-height:1.8;">A method is a function that belongs to a value, like a string.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">메서드는 문자열 같은 값이 가지고 있는 함수입니다.</p>

<h2 style="color:#8250df; font-size:1.7em; font-weight:700; margin-bottom:0.4em;"><code>upper()</code> / <code>lower()</code></h2>

<p style="line-height:1.8;"><code>upper()</code> makes uppercase. <code>lower()</code> makes lowercase.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;"><code>upper()</code>는 대문자, <code>lower()</code>는 소문자로 바꿉니다.</p>

```python
a = "Hello Python"
print(a.upper())
print(a.lower())
```

<h2 style="color:#1a7f37; font-size:1.7em; font-weight:700; margin-bottom:0.4em;"><code>strip()</code> / <code>lstrip()</code> / <code>rstrip()</code></h2>

<p style="line-height:1.8;"><code>strip()</code> removes spaces on both sides. <code>lstrip()</code> is left. <code>rstrip()</code> is right.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;"><code>strip()</code>은 양쪽 공백, <code>lstrip()</code>은 왼쪽, <code>rstrip()</code>은 오른쪽입니다.</p>

```python
a = "   Hello Python   "
print(a.strip())
print(a.lstrip())
print(a.rstrip())
```

<h2 style="color:#0550ae; font-size:1.7em; font-weight:700; margin-bottom:0.4em;"><code>isOO()</code> : check the string</h2>

<p style="line-height:1.8;">These methods return <code>True</code> or <code>False</code>.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">이 메서드들은 <code>True</code> 또는 <code>False</code>를 반환합니다.</p>

| Method | Meaning |
|--------|---------|
| <code>isalnum()</code> | letters or numbers only (문자 또는 숫자만) |
| <code>isalpha()</code> | letters only (문자만) |
| <code>isidentifier()</code> | a valid variable name (변수 이름으로 쓸 수 있음) |
| <code>isdecimal()</code> | decimal digits 0-9 (0-9 십진 숫자) |
| <code>isdigit()</code> | digits (숫자) |
| <code>isspace()</code> | whitespace only (공백만) |
| <code>islower()</code> | all lowercase (모두 소문자) |
| <code>isupper()</code> | all uppercase (모두 대문자) |

```python
print("abc123".isalnum())
print("abc".isalpha())
print("hello".isidentifier())
print("123".isdecimal())
print("123".isdigit())
print("   ".isspace())
print("hello".islower())
print("HELLO".isupper())
```

<h2 style="color:#8250df; font-size:1.7em; font-weight:700; margin-bottom:0.4em;"><code>find()</code> / <code>rfind()</code></h2>

<p style="line-height:1.8;"><code>find()</code> looks from the left. <code>rfind()</code> looks from the right. Both return an index. If not found, they return <code>-1</code>.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;"><code>find()</code>는 왼쪽부터, <code>rfind()</code>는 오른쪽부터 찾습니다. 인덱스를 반환하고, 없으면 <code>-1</code>입니다.</p>

```python
a = "안녕안녕"
print(a.find("안녕"))
print(a.rfind("안녕"))
print(a.find("파이썬"))
```

<h2 style="color:#1a7f37; font-size:1.7em; font-weight:700; margin-bottom:0.4em;"><code>in</code> operator</h2>

<p style="line-height:1.8;"><code>in</code> checks if a string is inside another string. It returns <code>True</code> or <code>False</code>.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;"><code>in</code>은 어떤 문자열이 안에 있는지 확인합니다. <code>True</code> 또는 <code>False</code>입니다.</p>

```python
print("안녕" in "안녕하세요")
print("잘자" in "안녕하세요")
```

<h2 style="color:#0550ae; font-size:1.7em; font-weight:700; margin-bottom:0.4em;"><code>split()</code></h2>

<p style="line-height:1.8;"><code>split()</code> cuts a string and returns a <strong>list</strong>.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;"><code>split()</code>은 문자열을 자르고 <strong>리스트</strong>를 반환합니다.</p>

```python
a = "10 20 30 40 50"
print(a.split())
print(type(a.split()))
```

<h2 style="color:#0969da; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">f-string</h2>

<p style="line-height:1.8;">Put an <code>f</code> in front of the string, and write a variable inside <code>{}</code>.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">문자열 앞에 <code>f</code>를 붙이고, <code>{}</code> 안에 변수를 넣습니다.</p>

<p style="line-height:1.8;"><code>format()</code> and f-string can do the same job. Use f-string when the code is short and you want to read it easily.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;"><code>format()</code>과 f-string은 같은 일을 할 수 있습니다. 코드가 짧고 읽기 쉬울 때는 f-string을 쓰세요.</p>

<p style="line-height:1.8;">You can put a variable or an expression right inside <code>{}</code>, so you do not need <code>.format(name, n)</code> at the end.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;"><code>{}</code> 안에 변수나 계산식을 바로 넣을 수 있어서, 뒤에 <code>.format(name, n)</code>을 붙이지 않아도 됩니다.</p>

<p style="line-height:1.8;">Use <code>format()</code> when you store a template first, then fill it later. Example: <code>"Hello, {}".format(name)</code>.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">문구를 먼저 저장해 두고 나중에 값을 넣을 때는 <code>format()</code>이 편합니다. 예: <code>"Hello, {}".format(name)</code>.</p>

```python
name = "KOS"
n = 10
print(f"My name is {name}.")
print(f"{n} + {n} = {n + n}")
print(f"{52:d}")
print(f"{52.273:.3f}")
```

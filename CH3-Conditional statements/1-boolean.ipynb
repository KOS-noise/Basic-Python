Boolean Operator

Bool has only two values, True and False.

We can make Bool by using comparison operator. 
Python has six comparison operators. 

(표로 작성)
- == : 같다
- != : 다르다
- < : 작다
- > : 크다 
- <= 작거나 같다
- >= 크거나 같다 

And also, Python provide alphabet comparison operator. In korean language Hanguel, items that come earlier in lexicographical (alphabetical) order have a smaller value.

Let's see

```python
print("가방" == "가방")

print("가방" < "하마")

print("graphics" > "genson")
```

```python
# 범위 구하기
x = 25
print(10 < x < 30)
print(40 < x < 60)
```

How to use date/time

```python
# call about function that associated in date/time
import datetime

# find a date/time in now
now = datetime.datetime.now()

# print

print(now.year, "년")
print(now.month, "월")
print(now.day, "일")
print(now.hour, "시")
print(now.minute, "분")
print(now.second, "초")
```

Before we studied, We can print the date clearly by using 'format()'.


```python
import datetime

now = datetime.datetime.now()

print("{}년 {}월 {}일 {}시 {}분 {}초".format(
    now.year,
    now.month,
    now.day,
    now.hour,
    now.minute,
    now.second
))
```

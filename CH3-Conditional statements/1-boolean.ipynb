<h1 style="color:#0969da; font-size:2.4em; font-weight:700; letter-spacing:-0.02em; margin-bottom:0.35em;">Boolean</h1>

<p style="line-height:1.8;">A bool has only two values: <code>True</code> and <code>False</code>.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">불은 <code>True</code>와 <code>False</code> 두 값만 가집니다.</p>

<h2 style="color:#8250df; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Comparison Operator</h2>

<p style="line-height:1.8;">We can make a bool with a comparison operator. Python has six comparison operators.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">비교 연산자로 불을 만들 수 있습니다. 파이썬에는 비교 연산자가 여섯 개 있습니다.</p>

| Operator | Meaning |
|----------|---------|
| <code>==</code> | equal (같다) |
| <code>!=</code> | not equal (다르다) |
| <code>&lt;</code> | less than (작다) |
| <code>&gt;</code> | greater than (크다) |
| <code>&lt;=</code> | less than or equal (작거나 같다) |
| <code>&gt;=</code> | greater than or equal (크거나 같다) |

<p style="line-height:1.8;">Python can also compare strings. In Korean, a word that comes earlier in dictionary order has a smaller value.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">파이썬은 문자열도 비교할 수 있습니다. 한글은 사전에서 앞에 오는 글자가 더 작습니다.</p>

```python
print("가방" == "가방")
print("가방" < "하마")
print("graphics" > "genson")
```

```python
# 범위 구하기
x = 25
print(10 < x < 30)
print(40 < x < 60)
```

<h2 style="color:#1a7f37; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Logical Operator</h2>

<p style="line-height:1.8;">We make bools with comparison operators. Then we can combine those bools with logical operators. Python has three logical operators.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">비교 연산자로 불을 만들고, 그 불들을 논리 연산자로 결합할 수 있습니다. 파이썬의 논리 연산자는 세 가지입니다.</p>

| Operator | Meaning |
|----------|---------|
| <code>not</code> | reverse True/False (아니다) |
| <code>and</code> | True only if both are True (그리고) |
| <code>or</code> | True if at least one is True (또는) |

<p style="line-height:1.8;"><code>not</code> is a unary operator. It takes one value.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;"><code>not</code>은 단항 연산자입니다. 값이 하나면 됩니다.</p>

<p style="line-height:1.8;"><code>and</code> and <code>or</code> are binary operators. They take two values.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;"><code>and</code>와 <code>or</code>는 이항 연산자입니다. 값이 두 개 필요합니다.</p>

```python
# not Operator
print(not True)
print(not False)
print(True and True)
print(True or False)
```

<h2 style="color:#0550ae; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">if statement</h2>

<p style="line-height:1.8;">An <code>if</code> statement runs code only when the condition is <code>True</code>.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;"><code>if</code> 조건문은 조건이 <code>True</code>일 때만 코드를 실행합니다.</p>

```python
number = input("정수 입력> ")
number = int(number)

if number > 0:
    print("positive")
else:
    print("negative")
```

<h2 style="color:#8250df; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Date and time</h2>

<p style="line-height:1.8;">The <code>datetime</code> module can get the current date and time.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;"><code>datetime</code> 모듈로 현재 날짜와 시간을 가져올 수 있습니다.</p>

```python
import datetime

now = datetime.datetime.now()

print(now.year, "년")
print(now.month, "월")
print(now.day, "일")
print(now.hour, "시")
print(now.minute, "분")
print(now.second, "초")
```

<p style="line-height:1.8;">We can print the date in one line with <code>format()</code>.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;"><code>format()</code>을 쓰면 날짜를 한 줄로 출력할 수 있습니다.</p>

```python
import datetime

now = datetime.datetime.now()

print("{}년 {}월 {}일 {}시 {}분 {}초".format(
    now.year,
    now.month,
    now.day,
    now.hour,
    now.minute,
    now.second
))
```

<p style="line-height:1.8;">An f-string puts the value inside <code>{}</code>. Do not leave <code>{}</code> empty.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">f-string은 <code>{}</code> 안에 값을 넣습니다. <code>{}</code>를 비우면 안 됩니다.</p>

```python
import datetime

now = datetime.datetime.now()
print(f"{now.year}년 {now.month}월 {now.day}일 {now.hour}시 {now.minute}분 {now.second}초")
```

```python
# 오전과 오후를 구분하는 프로그램
import datetime

now = datetime.datetime.now()

if now.hour < 12:
    print(f"현재 시각은 {now.hour}시로 오전입니다.")

if now.hour >= 12:
    print(f"현재 시각은 {now.hour}시로 오후입니다.")
```

```python
# 계절을 구분하는 프로그램
import datetime

now = datetime.datetime.now()

if 3 <= now.month <= 5:
    print(f"지금은 {now.month}월로 봄이다.")
if 6 <= now.month <= 8:
    print(f"지금은 {now.month}월로 여름이다.")
if 9 <= now.month <= 11:
    print(f"지금은 {now.month}월로 가을이다.")
if now.month == 12 or 1 <= now.month <= 2:
    print(f"지금은 {now.month}월로 겨울이다.")
```

<h2 style="color:#d1242f; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Even or odd</h2>

<p style="line-height:1.8;">We can check the last digit of a number.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">숫자의 마지막 자리로 짝수와 홀수를 구분할 수 있습니다.</p>

```python
# 끝자리로 짝수와 홀수 구분
number = input("정수 입력> ")

last_character = number[-1]
last_number = int(last_character)

if last_number == 0 \
    or last_number == 2 \
    or last_number == 4 \
    or last_number == 6 \
    or last_number == 8:
    print("짝수이다.")
else:
    print("홀수이다.")
```

```python
# in 문자열 연산자 활용해서 짝수와 홀수 구분
number = input("정수 입력> ")
last_character = number[-1]

if last_character in "02468":
    print("짝수이다.")
else:
    print("홀수이다.")
```

```python
# 나머지 연산자를 활용해서 짝수와 홀수 구분
number = input("정수 입력> ")
number = int(number)

if number % 2 == 0:
    print("짝수이다.")
else:
    print("홀수이다.")
```

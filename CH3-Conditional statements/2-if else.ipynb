<h1 style="color:#0969da; font-size:2.4em; font-weight:700; letter-spacing:-0.02em; margin-bottom:0.35em;">if ~ else and elif</h1>

<p style="line-height:1.8;"><code>if</code> runs when the condition is <code>True</code>. <code>else</code> runs when it is <code>False</code>.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;"><code>if</code>는 조건이 <code>True</code>일 때 실행합니다. <code>else</code>는 <code>False</code>일 때 실행합니다.</p>

```python
number = input("정수 입력> ")
number = int(number)

if number > 0:
    print("positive")
else:
    print("negative or zero")
```

<h2 style="color:#8250df; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">elif</h2>

<p style="line-height:1.8;"><code>elif</code> means “else if”. Use it when there are more than two cases.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;"><code>elif</code>는 “아니면 만약”입니다. 경우가 세 개 이상일 때 씁니다.</p>

<p style="line-height:1.8;">Python checks from the top. When one condition is <code>True</code>, it skips the rest.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">위에서부터 확인하고, 하나가 <code>True</code>면 나머지는 건너뜁니다.</p>

```python
number = input("정수 입력> ")
number = int(number)

if number > 0:
    print("positive")
elif number < 0:
    print("negative")
else:
    print("zero")
```

<h2 style="color:#1a7f37; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Values that become False</h2>

<p style="line-height:1.8;">If the <code>if</code> condition is not a bool, Python converts it to a bool first.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;"><code>if</code> 조건에 불이 아닌 값이 오면, 파이썬이 먼저 불로 바꿉니다.</p>

<p style="line-height:1.8;">These values become <code>False</code>:</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;"><code>False</code>로 바뀌는 값:</p>

- <code>None</code>
- <code>0</code> and <code>0.0</code>
- empty containers: <code>""</code>, <code>b""</code>, <code>[]</code>, <code>()</code>, <code>{}</code>

<p style="line-height:1.8;">Everything else becomes <code>True</code>. Remember only these three kinds.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">이 외에는 모두 <code>True</code>입니다. 위 세 가지만 기억하면 됩니다.</p>

```python
print("False values")
if 0:
    print("0 is True")
else:
    print("0 is False")

if 0.0:
    print("0.0 is True")
else:
    print("0.0 is False")

if None:
    print("None is True")
else:
    print("None is False")

if "":
    print("empty string is True")
else:
    print("empty string is False")

if []:
    print("empty list is True")
else:
    print("empty list is False")

print()
print("True values")
if 1:
    print("1 is True")
if "Hello":
    print("Hello is True")
```

<h2 style="color:#0550ae; font-size:1.7em; font-weight:700; margin-bottom:0.4em;"><code>pass</code></h2>

<p style="line-height:1.8;"><code>pass</code> means “do nothing”. Use it when the block is not ready yet.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;"><code>pass</code>는 “아무것도 하지 않음”입니다. 아직 코드를 안 썼을 때 씁니다.</p>

```python
number = 10

if number > 0:
    pass
else:
    pass

print("no error")
```

<h2 style="color:#8250df; font-size:1.7em; font-weight:700; margin-bottom:0.4em;"><code>raise NotImplementedError</code></h2>

<p style="line-height:1.8;"><code>pass</code> hides that the code is empty. <code>raise NotImplementedError</code> shows it clearly: “this part is not written yet”.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;"><code>pass</code>는 빈 코드를 조용히 넘어갑니다. <code>raise NotImplementedError</code>는 “아직 안 만들었다”고 분명히 알려 줍니다.</p>

<p style="line-height:1.8;">The next cell will raise an error on purpose.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">다음 셀은 일부러 오류가 납니다.</p>

```python
number = 10

if number > 0:
    raise NotImplementedError
else:
    raise NotImplementedError
```

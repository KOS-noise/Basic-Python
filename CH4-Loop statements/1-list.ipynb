<h1 style="color:#0969da; font-size:2.4em; font-weight:700; letter-spacing:-0.02em; margin-bottom:0.35em;">List</h1>

<p style="line-height:1.8;">A list stores several values in one place. The values can have different data types.</p>

<p style="line-height:1.8;">So far we used one value at a time: a string, a number, or a bool. A list collects those values together.</p>

```python
array = [273, 32, 103, "String", True, False]

array
```

<h2 style="color:#8250df; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Element</h2>

<p style="line-height:1.8;">Put values inside square brackets <code>[]</code>. Each value is called an <strong>element</strong>(요소).</p>

<p style="line-height:1.8;">Separate elements with a comma <code>,</code>.</p>

<h2 style="color:#1a7f37; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Index</h2>

<p style="line-height:1.8;">An <strong>index</strong>(인덱스) is the position of an element. Python starts at <code>0</code>.</p>

| Index | 0 | 1 | 2 | 3 | 4 | 5 |
|-------|---|---|---|---|---|---|
| Element | <code>273</code> | <code>32</code> | <code>103</code> | <code>"String"</code> | <code>True</code> | <code>False</code> |

<p style="line-height:1.8;">Use <code>list[index]</code> to get one element.</p>

```python
list_a = [273, 32, 103, "String", True, False]
```

```python
list_a[0]
```

<h2 style="color:#0550ae; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Slice</h2>

<p style="line-height:1.8;">A slice(슬라이스) gets a range of elements: <code>list[start:end]</code>.</p>

<p style="line-height:1.8;"><code>start</code> is included. <code>end</code> is not included. So <code>list_a[1:3]</code> is index 1 and 2.</p>

```python
list_a[1:3]
```

<p style="line-height:1.8;">We can also change one element with <code>=</code>.</p>

```python
# change one element
list_a[0] = "change"
list_a
```

<h2 style="color:#d1242f; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Negative index</h2>

<p style="line-height:1.8;"><code>-1</code> is the last element. <code>-2</code> is the second from the last.</p>

| Index | -6 | -5 | -4 | -3 | -2 | -1 |
|-------|----|----|----|----|----|----|
| Element | <code>"change"</code> | <code>32</code> | <code>103</code> | <code>"String"</code> | <code>True</code> | <code>False</code> |

```python
list_a[-1]
list_a[-3]
```

<p style="line-height:1.8;">A notebook cell shows only the last line. Use <code>print()</code> to see both.</p>

```python
print(list_a[-1])
print(list_a[-3])
```

<h2 style="color:#8250df; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Index twice</h2>

<p style="line-height:1.8;">If an element is a string, we can use the index operator twice.</p>

<p style="line-height:1.8;"><code>list_a[3]</code> is <code>"String"</code>. Then <code>list_a[3][0]</code> is the first character, <code>"S"</code>.</p>

```python
list_a[3][0]
```

<h2 style="color:#1a7f37; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">List in a list</h2>

<p style="line-height:1.8;">A list can contain another list. This is a nested list(중첩 리스트).</p>

<p style="line-height:1.8;"><code>list_b[0]</code> is the first inner list. <code>list_b[1][1]</code> is the second element of the second inner list.</p>

```python
list_b = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
print(list_b[0])
print(list_b[1][1])
```

<h2 style="color:#0550ae; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">IndexError</h2>

<p style="line-height:1.8;">If the index is out of range, Python raises an <strong>IndexError</strong>. This is the same error as with strings.</p>

<p style="line-height:1.8;">The next cell will raise an error on purpose. That is normal.</p>

```python
list_a[100]
```

<h1 style="color:#0969da; font-size:2.4em; font-weight:700; letter-spacing:-0.02em; margin-bottom:0.35em;">List operators</h1>

<p style="line-height:1.8;">Lists have operators, like strings. The three basic ones are concatenate, repeat, and length.</p>

| Operator | Meaning |
|----------|---------|
| <code>+</code> | concatenate (연결) |
| <code>*</code> | repeat (반복) |
| <code>len()</code> | length (길이) |

```python
list_a = [1, 2, 3]
list_b = [4, 5, 6]

print(list_a + list_b)
print(list_a * 3)
print(len(list_a))
```

<p style="line-height:1.8;"><code>list_a + list_b</code> makes a <strong>new</strong> list. It does not change <code>list_a</code> or <code>list_b</code>.</p>

```python
print(list_a)
print(list_b)
```

<h2 style="color:#8250df; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Add an element</h2>

<p style="line-height:1.8;">There are three common methods. All of them change the list in place.</p>

| Method | Meaning |
|--------|---------|
| <code>append(value)</code> | add at the end |
| <code>insert(index, value)</code> | add at that index |
| <code>extend(list)</code> | add many values at the end |

<p style="line-height:1.8;"><code>append()</code> always goes to the end. <code>insert()</code> can go to any index.</p>

```python
list_a = [1, 2, 3]
list_b = [4, 5, 6]

list_a.append(4)
print("append:", list_a)

list_b.insert(1, "insert")
print("insert:", list_b)

list_a.extend([5, 6, 7])
print("extend:", list_a)
```

<p style="line-height:1.8;"><code>+</code> does not change the original list. <code>extend()</code> does.</p>

```python
list_a = [1, 2, 3]
list_b = [4, 5, 6]

print(list_a + list_b)
print("after + :", list_a)

list_a.extend(list_b)
print("after extend():", list_a)
```

<h2 style="color:#1a7f37; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Remove an element</h2>

<p style="line-height:1.8;">There are two main ways: remove by <strong>index</strong>, or remove by <strong>value</strong>.</p>

| How | Code |
|-----|------|
| by index | <code>del list[index]</code> |
| by index | <code>list.pop(index)</code> |
| by range | <code>del list[start:end]</code> |
| by value | <code>list.remove(value)</code> |
| all | <code>list.clear()</code> |

<h2 style="color:#0550ae; font-size:1.7em; font-weight:700; margin-bottom:0.4em;"><code>del</code> and <code>pop()</code></h2>

<p style="line-height:1.8;"><code>del</code> removes the element at that index. <code>pop()</code> also removes it, but it <strong>returns</strong> the value.</p>

<p style="line-height:1.8;"><code>pop()</code> with no argument removes the last element.</p>

```python
list_a = [1, 2, 3, 4, 5, 6, 7]

del list_a[3]
print("after del [3]:", list_a)

last = list_a.pop()
print("pop():", last)
print("after pop():", list_a)

list_a.pop(1)
print("after pop(1):", list_a)
```

<p style="line-height:1.8;"><code>del</code> can also remove a range, like a slice.</p>

```python
list_b = [4, "insert", 5, 6]

del list_b[:3]
print(list_b)
```

<h2 style="color:#d1242f; font-size:1.7em; font-weight:700; margin-bottom:0.4em;"><code>remove()</code> and <code>clear()</code></h2>

<p style="line-height:1.8;"><code>remove(value)</code> finds the first matching value and deletes it. You do not use an index.</p>

<p style="line-height:1.8;"><code>clear()</code> removes every element.</p>

```python
list_c = [1, 2, 3, 4]
list_c.remove(3)
print("after remove(3):", list_c)

list_c.clear()
print("after clear():", list_c)
```

<h2 style="color:#8250df; font-size:1.7em; font-weight:700; margin-bottom:0.4em;"><code>sort()</code> and <code>reverse()</code></h2>

<p style="line-height:1.8;"><code>sort()</code> sorts the list in place. The default is ascending(오름차순). Use <code>reverse=True</code> for descending(내림차순).</p>

<p style="line-height:1.8;"><code>reverse()</code> only flips the current order. It does not sort.</p>

```python
list_c = [1, 4, 2]

list_c.sort()
print("sort():", list_c)

list_c.sort(reverse=True)
print("sort(reverse=True):", list_c)

list_d = [1, 2, 3, 4, 5]
list_d.reverse()
print("reverse():", list_d)
```

<h2 style="color:#1a7f37; font-size:1.7em; font-weight:700; margin-bottom:0.4em;"><code>in</code> / <code>not in</code></h2>

<p style="line-height:1.8;">Use <code>in</code> to check if a value is inside a list. It returns <code>True</code> or <code>False</code>.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">리스트 내부에 있는지 확인하기: in/not in 연산자</p>

<p style="line-height:1.8;"><code>not in</code> is the opposite. It is <code>True</code> when the value is not in the list.</p>

| Operator | Meaning |
|----------|---------|
| <code>value in list</code> | True if the value is in the list |
| <code>value not in list</code> | True if the value is not in the list |

```python
list_e = [273, 32, 103, 57, 52]

print(273 in list_e)
print(99 in list_e)
print(273 not in list_e)
print(99 not in list_e)
```

<h1 style="color:#0969da; font-size:2.4em; font-weight:700; letter-spacing:-0.02em; margin-bottom:0.35em;">Dictionary</h1>

<p style="line-height:1.8;">A list stores values by index. A dictionary stores values by <strong>key</strong>.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">리스트는 인덱스로 값을 저장한다. 딕셔너리는 키로 값을 저장한다.</p>

<p style="line-height:1.8;">The basic form is:</p>

```python
{
    "key": value
}
```

<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;"><code>{ "키": 값 }</code></p>

<h2 style="color:#8250df; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Declare a dictionary</h2>

<p style="line-height:1.8;">A dictionary uses curly brackets <code>{}</code>. Write each pair as <code>key: value</code>, and separate pairs with a comma <code>,</code>.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">딕셔너리는 중괄호 <code>{}</code>로 선언한다. <code>key: value</code> 형태를 쉼표로 연결한다.</p>

<p style="line-height:1.8;">A key can be a string, a number, a bool, and so on.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">키에는 문자열, 숫자, 불 등을 넣을 수 있다.</p>

````python
dict_a = {
    'name': '어벤져스 엔드게임',
    'type': '히어로 무비'
}
````

````python
dict_a
````

````python
dict_a['name']
````

<p style="line-height:1.8;">A value can be a string, a number, or a bool. A list and a dictionary are also values, so a value can be a list or another dictionary.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">값에는 문자열, 숫자, 불을 넣을 수 있다. 리스트와 딕셔너리도 값이므로, 값 자리에 리스트나 딕셔너리를 넣을 수 있다.</p>

````python
dict_b = {
    'director': ['안소니 루소', '조 루소'],
    'cast': ['아이언맨', '타노스', '토르', '닥터스트레인지', '헐크']
}
````

````python
dict_b
````

<h2 style="color:#1a7f37; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Access a dictionary element</h2>

<p style="line-height:1.8;">Use <code>dictionary[key]</code> to get the value for that key.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">딕셔너리의 요소에 접근하기</p>

````python
dictionary = {
    'name': '참 크래커',
    'type': '과자',
    'ingredient': ['탄수화물', '단백질', '지방', '나트륨'],
    'origin': '한국'
}
````

````python
print('name:', dictionary["name"])
print('type:', dictionary["type"])
print('ingredient:', dictionary["ingredient"])
print('origin:', dictionary["origin"])
print()
````

````python
# change the value

dictionary['name'] = 'Cham Cracker'
print('name:', dictionary['name'])
````

<h2 style="color:#d1242f; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">NameError</h2>

<p style="line-height:1.8;">If you write a key with no quotes, Python reads it as a variable name. If that name does not exist, you get a <strong>NameError</strong>.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">파이썬은 딕셔너리의 키에 따옴표 없이 단순하게 식별자를 입력하면 이를 변수로 인식합니다.</p>

<p style="line-height:1.8;"><code>dictionary["name"]</code> uses the key <code>"name"</code>. <code>dictionary[name]</code> looks for a variable called <code>name</code>.</p>

<h2 style="color:#0550ae; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Add a value</h2>

<p style="line-height:1.8;">Assign a value to a new key. If the key is new, Python adds that pair.</p>

```python
dictionary[new_key] = new_value
```

<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;"><code>딕셔너리[새로운 키] = 새로운 값</code></p>

````python
dictionary['price'] = 1200
dictionary
````

````python
# delete element

del dictionary['ingredient']

dictionary
````

<h2 style="color:#d1242f; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">KeyError</h2>

<p style="line-height:1.8;">If the key is not in the dictionary, <code>dictionary[key]</code> raises a <strong>KeyError</strong>.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">딕셔너리에 없는 키로 접근하면 <strong>KeyError</strong>가 난다.</p>

<h2 style="color:#8250df; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Check a key with <code>in</code></h2>

<p style="line-height:1.8;"><code>key in dictionary</code> is <code>True</code> when that key exists. Use it before you access the key, so you can avoid a KeyError.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">딕셔너리 내부에 키가 있는지 확인하기 : in 키워드</p>

````python
key = input(">접근하고자 하는 키: ")
````

````python
if key in dictionary:
    print(dictionary[key])
else:
    print("존재하지 않는 키에 접근하고 있습니다. ")
````

<h2 style="color:#1a7f37; font-size:1.7em; font-weight:700; margin-bottom:0.4em;"><code>get()</code></h2>

<p style="line-height:1.8;"><code>get()</code> returns the value for a key. If the key does not exist, it returns <code>None</code>. It does not raise a KeyError.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">get() 함수는 딕셔너리의 키로 값을 추출하는 기능으로 존재하지 않는 키에 접근할 경우 KeyError 를 발생시키지 않고 None을 출력한다.</p>

````python
value = dictionary.get("존재하지 않는 키")
print('값:', value)
````

````python
# None 확인 방법
if value == None:
    print("존재하지 않는 키에 접근했었습니다.")
````

<h2 style="color:#0550ae; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">for loop with a dictionary</h2>

<p style="line-height:1.8;">A <code>for</code> loop on a dictionary gives the <strong>keys</strong>. Use <code>dictionary[key]</code> to get each value.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">for 반복문: 딕셔너리와 함께 사용하기</p>

```python
for key in dictionary:
    code
```

<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;"><code>for 키 변수 in 딕셔너리:</code><br>&nbsp;&nbsp;&nbsp;&nbsp;<code>코드</code></p>

````python
# for 반복문 사용
for key in dictionary:
    print(key, ':', dictionary[key])
````

<h1 style="color:#0969da; font-size:2.4em; font-weight:700; letter-spacing:-0.02em; margin-bottom:0.35em;">range</h1>

<p style="line-height:1.8;"><code>range</code> is often used with a <code>for</code> loop. Use them together when you want to repeat a specific number of times.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">범위 자료형과 while 반복문</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">for 반복문과 함께 많이 사용되는 자료형은 범위 range 가 있다. 특정한 횟수만큼 반복해서 돌리고 싶을 때는 for 반복문과 범위를 조합해서 사용한다.</p>

<p style="line-height:1.8;">There are three ways to call <code>range()</code>.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">범위는 다음과 같은 3가지 방법으로 사용할 수 있다.</p>

- one number: <code>range(A)</code>
- two numbers: <code>range(A, B)</code>
- three numbers: <code>range(A, B, C)</code>

<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">매개변수에 숫자 한 개를 넣는 방법<br>매개변수에 숫자 두 개를 넣는 방법<br>매개변수에 숫자 세 개를 넣는 방법</p>

<h2 style="color:#8250df; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">One argument</h2>

<p style="line-height:1.8;"><code>range(A)</code> makes integers from <code>0</code> to <code>A - 1</code>.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">첫째, 매개변수에 숫자를 한 개 넣는 방법</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">0부터 A-1 까지의 정수로 범위를 만든다.</p>

```python
range(A)
```

````python
a = range(5)
print(a)
print(list(a))
````

<h2 style="color:#1a7f37; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Two arguments</h2>

<p style="line-height:1.8;"><code>range(A, B)</code> makes integers from <code>A</code> to <code>B - 1</code>. <code>B</code> is not included.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">둘째, 매개변수에 숫자를 두 개 넣는 방법</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">A부터 B-1 까지의 정수로 범위를 만든다.</p>

```python
range(A, B)
```

````python
print(list(range(2, 6)))
````

<h2 style="color:#0550ae; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Three arguments</h2>

<p style="line-height:1.8;"><code>range(A, B, C)</code> also goes from <code>A</code> to <code>B - 1</code>. Each number is <code>C</code> apart. <code>C</code> is the step.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">셋째, 매개변수에 숫자를 세 개 넣는 방법</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">A부터 B-1까지의 정수로 범위를 만드는데, 앞뒤의 숫자가 C만큼의 차이를 가진다.</p>

```python
range(A, B, C)
```

````python
print(list(range(0, 10, 2)))
````

<h2 style="color:#d1242f; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">TypeError</h2>

<p style="line-height:1.8;">Every argument of <code>range()</code> must be an integer. <code>/</code> makes a float, so the next cell raises a <strong>TypeError</strong>. That is normal.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">TypeError 가 발생하는 이유는 range() 함수의 매개변수로는 반드시 '정수'를 입력해야 하기 때문이다.</p>

````python
#TypeError

n = 10

b = range(0, n/2)
````

````python
b = range(0, n//2) # 정수 나누기 연산자를 많이 사용함
b
````

<h2 style="color:#8250df; font-size:1.7em; font-weight:700; margin-bottom:0.4em;"><code>for</code> loop with <code>range</code></h2>

<p style="line-height:1.8;">The basic form is:</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">for 반복문: 범위</p>

```python
for number in range(...):
    code
```

<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;"><code>for 숫자 변수 in 범위:</code><br>&nbsp;&nbsp;&nbsp;&nbsp;<code>코드</code></p>

````python
for i in range(5):
    print(i)
````

<h2 style="color:#1a7f37; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">List and <code>range</code> together</h2>

<p style="line-height:1.8;">Sometimes we need the index: which repetition is this? <code>range(len(list))</code> gives <code>0, 1, 2, ...</code> for that list.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">for 반복문: 리스트와 범위 조합하기</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">반복을 적용하다 보면 '몇 번째 반복일까?'를 알아야 하는 경우가 있다.</p>

````python
array = [273, 32, 103, 57, 52]

for i in range(len(array)):
    print("{}번째 반복: {}".format(i, array[i]))
````

<h2 style="color:#0550ae; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Reverse loop</h2>

<p style="line-height:1.8;">There are two common ways to loop backward.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">for 반복문: 반대로 반복하기</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">역반복문은 크게 두 가지 방법을 사용해서 만든다.</p>

<p style="line-height:1.8;">The first way uses three arguments. A negative step goes backward. <code>range(4, -1, -1)</code> is <code>4, 3, 2, 1, 0</code>. In the next cell, <code>0 - 1</code> is <code>-1</code>.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">첫번째는 range() 함수의 매개변수를 세 개 사용하는 방법이 있다.</p>

````python
for i in range(4, 0-1, -1):
    print("현재 반복 변수: {}".format(i))
````

<p style="line-height:1.8;">The second way uses <code>reversed()</code>. It flips the order of a range.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">두 번째 방법은 reversed() 함수를 사용하는 것이 있다.</p>

````python
# 역반복문
for i in reversed(range(5)):
    print("현재 반복 변수: {}".format(i))
````

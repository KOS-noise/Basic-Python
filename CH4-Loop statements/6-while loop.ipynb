<h1 style="color:#0969da; font-size:2.4em; font-weight:700; letter-spacing:-0.02em; margin-bottom:0.35em;">while loop</h1>

<p style="line-height:1.8;">A <code>while</code> loop repeats the block while the condition is <code>True</code>.</p>

```python
while condition:
    statement
```

<p style="line-height:1.8;"><code>while True</code> never becomes <code>False</code>, so the loop does not stop. This is an infinite loop.</p>

<p style="line-height:1.8;"><code>end=""</code> prints the next value on the same line. Stop the cell with the interrupt button if you run the next cell.</p>

````python
# 무한 반복
while True:
    print(".", end = "")
````

<h2 style="color:#8250df; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Use <code>while</code> like <code>for</code></h2>

<p style="line-height:1.8;">A <code>for</code> loop can repeat a fixed number of times with <code>range</code>. A <code>while</code> loop can do the same if we count with a variable and stop when the condition is <code>False</code>.</p>

````python
i = 0

while i < 10:
    print("{}번째 반복입니다".format(i))
    i += 1
````

<p style="line-height:1.8;">The condition can also check a list. This loop removes <code>value</code> while that value is still in the list.</p>

````python
list_test = [1, 2, 1, 2]
value = 2

# list_test 내부에 value 가 있다면 반복한다. 
while value in list_test:
    list_test.remove(value)

print(list_test)
````

<h2 style="color:#1a7f37; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Repeat for a length of time</h2>

<p style="line-height:1.8;">To repeat for a number of seconds, we use <strong>Unix time</strong>.</p>

<p style="line-height:1.8;">Unix time is the number of seconds since 1970-01-01 00:00:00 UTC. In Python, <code>time.time()</code> returns that number.</p>

<p style="line-height:1.8;">First, import the time module.</p>

````python
import time
````

````python
time.time()
````

<p style="line-height:1.8;"><code>target_tick</code> is the current Unix time plus 5 seconds. The loop runs while <code>time.time()</code> is still earlier than that.</p>

<p style="line-height:1.8;">The next cell takes about 5 seconds.</p>

````python
number = 0

# 5초 동안 반복한다. 
target_tick = time.time() + 5
while time.time() < target_tick:
    number += 1

print("5초 동안 {}번 반복했습니다".format(number))
````

<p style="line-height:1.8;">The count changes with the computer and what else it is doing. The same pattern can wait a few seconds for another user's reply. When the condition is based on time, use a <code>while</code> loop.</p>

<h2 style="color:#0550ae; font-size:1.7em; font-weight:700; margin-bottom:0.4em;"><code>break</code> and <code>continue</code></h2>

<p style="line-height:1.8;"><code>break</code> stops the loop immediately.</p>

<p style="line-height:1.8;">The next cell asks for input on every repetition. Type <code>y</code> to leave the loop.</p>

````python
i = 0

while True:
    print("{}번째 반복문이다.".format(i))
    i = i + 1

    # 반복 종료
    input_text = input(">Do you wanna break? (y/n): ")

    if input_text in ["y", "Y"]:
        print("GOOD BYE")
        break
````

<p style="line-height:1.8;"><code>continue</code> skips the rest of the current iteration and starts the next one.</p>

<p style="line-height:1.8;">In the next cell, numbers less than 10 are skipped. Only 15, 20, and 25 are printed.</p>

````python
numbers = [5, 15, 6, 20, 7, 25]

for number in numbers:
    if number < 10:
        continue
    print(number)
````

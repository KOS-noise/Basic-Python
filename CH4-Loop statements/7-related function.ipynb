<h1 style="color:#0969da; font-size:2.4em; font-weight:700; letter-spacing:-0.02em; margin-bottom:0.35em;">Basic functions for lists and dictionaries</h1>

<p style="line-height:1.8;">These functions are often used with strings, lists, and dictionaries.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">문자열, 리스트, 딕셔너리와 관련된 기본 함수</p>

- <code>min()</code>, <code>max()</code>, <code>sum()</code>: work on a list
- <code>reversed()</code>: reverse a list
- <code>enumerate()</code>: get the current index
- <code>items()</code>: loop over a dictionary easily
- list comprehension: a <code>for</code> loop inside a list

<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">리스트에 적용할 수 있는 기본 함수: min(), max(), sum()<br>리스트 뒤집기: reversed()<br>현재 인덱스가 몇 번째인지 확인하기: enumerate()<br>딕셔너리로 쉽게 반복문 작성하기: items()<br>리스트 안에 for문 사용하기: 리스트 내포</p>

<h2 style="color:#8250df; font-size:1.7em; font-weight:700; margin-bottom:0.4em;"><code>min()</code>, <code>max()</code>, <code>sum()</code></h2>

<p style="line-height:1.8;"><code>min()</code> returns the smallest value. <code>max()</code> returns the largest. <code>sum()</code> adds the values.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;"><code>min()</code>은 가장 작은 값, <code>max()</code>는 가장 큰 값, <code>sum()</code>은 합이다.</p>

````python
numbers = [103, 52, 273, 32, 77]
````

````python
print(min(numbers))
print(max(numbers))
print(sum(numbers))
````

<p style="line-height:1.8;">These functions also take separate numbers. A list is not required.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">리스트를 쓰지 않고 숫자를 직접 넣어도 된다.</p>

````python
# 리스트를 사용하지 않을 때 이렇게 가능함
min(103, 52, 273)
````

<h2 style="color:#1a7f37; font-size:1.7em; font-weight:700; margin-bottom:0.4em;"><code>reversed()</code></h2>

<p style="line-height:1.8;"><code>reversed()</code> does not return a list. It returns a reverse <strong>iterator</strong>. An iterator gives the next value one by one.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;"><code>reversed()</code>는 리스트가 아니라 이터레이터를 반환한다. 이터레이터는 값을 하나씩 꺼낸다. 이 결과에 대해서는 이터레이터에서 좀 더 자세히 설명.</p>

<p style="line-height:1.8;">Use <code>list()</code> when you want a real list. A <code>for</code> loop can use the iterator directly.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;"><code>list()</code>로 감싸면 리스트가 된다. <code>for</code>에는 이터레이터를 바로 넣어도 된다.</p>

````python
list_a = [1, 2, 3, 4, 5]
list_reversed = reversed(list_a)

print("reversed([1, 2, 3, 4, 5])):", list_reversed) # 이 결과에 대해서는 이터레이터에서 좀 더 자세히 설명
print("list(reversed([1,2,3,4,5])):", list(list_reversed))
````

````python
# 반복문을 적용해 보면

for i in reversed(list_a):
    print("-", i)
````

<h2 style="color:#0550ae; font-size:1.7em; font-weight:700; margin-bottom:0.4em;"><code>enumerate()</code></h2>

<p style="line-height:1.8;">Start with this list.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">다음과 같은 리스트가 있다고 가정해 보겠습니다.</p>

````python
example_list = ["A", "B", "C"]
````

<p style="line-height:1.8;">We want the index and the value together:</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">다음과 같은 결과를 출력하고 싶다면 어떻게 해야 할까요?</p>

```text
0번째 요소는 A 입니다.
1번째 요소는 B 입니다.
2번째 요소는 C 입니다.
```

<p style="line-height:1.8;">One way is to count with a variable. Another way is <code>range(len(list))</code>.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">변수로 세거나, <code>range(len(list))</code>로 인덱스를 만들 수 있다.</p>

````python
# 방법 1

i = 0
for item in example_list:
    print("{}번째 요소는 {}입니다.".format(i, item))
    i += 1
````

````python
# 방법 2

for i in range(len(example_list)):
    print("{}번째 요소는 {}입니다.".format(i, example_list[i]))
````

<p style="line-height:1.8;">We often need the index while looping. <code>enumerate()</code> does that for us. It pairs each index with its value: <code>(0, "A")</code>, <code>(1, "B")</code>, <code>(2, "C")</code>.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">리스트의 요소를 반복할 때 현재 인덱스가 몇 번째인지 확인해야 하는 경우가 많다. 파이썬은 이런 코드를 쉽게 작성할 수 있도록 enumerate() 함수를 제공한다.</p>

<p style="line-height:1.8;">Like <code>reversed()</code>, <code>enumerate()</code> returns an iterator. <code>list()</code> shows the pairs.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;"><code>enumerate()</code>도 이터레이터다. <code>list()</code>로 감싸면 쌍이 보인다.</p>

````python
# enumerate() 함수와 리스트

example_list = ["A", "B", "C"]

print(enumerate(example_list))
print()
````

````python
# list() 함수로 강제 변환해 출력

print(list(enumerate(example_list)))
````

<p style="line-height:1.8;">With <code>enumerate()</code>, the <code>for</code> loop can take two variables: the index and the value.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">enumerate 함수를 사용하면 for와 in 사이에 반복 변수를 두 개 넣을 수 있습니다.</p>

````python
# for 반복문과 enumerate() 함수 조합해서 사용하기

for i, value in enumerate(example_list):
    print("{}번째 요소는 {} 입니다.".format(i, value))
````

<h2 style="color:#d1242f; font-size:1.7em; font-weight:700; margin-bottom:0.4em;"><code>items()</code></h2>

<p style="line-height:1.8;"><code>dict.items()</code> returns key-value pairs. A <code>for</code> loop can take both the key and the value, just like <code>enumerate()</code>.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">딕셔너리의 items() 함수와 반복문</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;"><code>items()</code>는 키와 값을 쌍으로 준다. <code>for</code>에서 변수를 두 개 받으면 된다.</p>

````python
# 변수 선언

example_dictionary = {
    "key1" : "value1",
    "key2" : "value2",
    "key3" : "value3"
}

# 딕셔너리의 items() 함수 결과 출력하기
print(example_dictionary.items())
````

````python
# for 반복문과 items() 함수 조합해서 사용하기

for key, element in example_dictionary.items():
    print("dictionary[{}] = {}".format(key, element))
````

<h2 style="color:#8250df; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">List comprehension</h2>

<p style="line-height:1.8;">We often use a loop to build a new list. The next cell takes even numbers from 0 to 18 with <code>range(0, 20, 2)</code>, squares each one, and appends it.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">리스트 내포</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">프로그램을 만들 때는 반복문을 사용해 리스트를 재조합하는 경우가 많다. 다음 코드는 range(0, 20, 2)로 0부터 20 사이의 짝수를 구한 뒤, 제곱해서 새로운 리스트를 만든다.</p>

````python
array = []

for i in range(0, 20, 2):
    array.append(i * i)

print(array)
````

<p style="line-height:1.8;">Python can write that loop in one line. Put the result expression first, then <code>for</code>.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">위 코드를 파이썬에서는 한 줄로 작성할 수 있습니다. 최종 결과를 맨 앞에 작성한다.</p>

````python
array2 = [i * i for i in range(0, 20, 2)] # 최종 결과를 맨 앞에 작성 
array2
````

<p style="line-height:1.8;">This form is a <strong>list comprehension</strong>. Add <code>if</code> at the end when you want a condition.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">이런 구문의 구조를 리스트 내포 (list comprehensions) 라고 부른다. 다음과 같이 if 뒤에 if 구문을 넣어 조건을 조합할 수도 있다.</p>

```python
[expression for item in iterable if condition]
```

<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;"><code>리스트 이름 = [표현식 for 반복자 in 반복할 수 있는 것 if 조건문]</code></p>

<p style="line-height:1.8;">The next cell keeps only numbers greater than 10, then squares them.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">다음 셀은 10보다 큰 수만 남겨서 제곱한다.</p>

````python
array3 = [i * i for i in range(0, 20, 2) if i > 10]
array3
````

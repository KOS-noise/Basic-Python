<h1 style="color:#0969da; font-size:2.4em; font-weight:700; letter-spacing:-0.02em; margin-bottom:0.35em;">Make a function</h1>

<p style="line-height:1.8;">A function is a named block of code. It runs when we call it.</p>

<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">함수는 이름이 있는 코드 블록입니다. 함수를 호출하면 실행됩니다.</p>
<p style="line-height:1.8;">Functions help us reuse code. They can also take input and return a result.</p>

<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">함수를 사용하면 코드를 다시 쓸 수 있습니다. 함수는 입력을 받고 결과를 돌려줄 수도 있습니다.</p>

<h2 style="color:#8250df; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Define and call a function</h2>

<p style="line-height:1.8;">Use <code>def</code> to define a function. The indented lines are its body.</p>

<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;"><code>def</code>를 사용해 함수를 정의합니다. 들여쓴 줄은 함수의 본문입니다.</p>

<pre><code>def function_name():
    code</code></pre>

<p style="line-height:1.8;">Add parentheses to call the function.</p>

<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">함수를 호출하려면 괄호를 붙입니다.</p>

```python
def say_hello():
    print("Hello!")

say_hello()
```

<h2 style="color:#1a7f37; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Parameters and arguments</h2>

<p style="line-height:1.8;">A <strong>parameter</strong> is a name in the function definition. It receives a value.</p>

<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">매개변수는 함수 정의에 적는 이름이며, 값을 전달받습니다.</p>
<p style="line-height:1.8;">An <strong>argument</strong> is a value we pass when we call the function.</p>

<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">인수는 함수를 호출할 때 전달하는 값입니다.</p>

<pre><code>def greet(name):  # name is a parameter
    print("Hello,", name)

greet("Mina")     # "Mina" is an argument</code></pre>

<p style="line-height:1.8;">Think of a parameter as an empty box. The argument is the value we put in that box.</p>

<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">매개변수는 빈 상자, 인수는 그 상자에 넣는 값이라고 생각하면 됩니다.</p>

```python
def greet(name):
    print("Hello,", name)

greet("Mina")
greet("Joon")
```

<h2 style="color:#0550ae; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Regular parameters and positional arguments</h2>

<p style="line-height:1.8;">A regular parameter (일반 매개변수) has no default value. The caller must give it a value.</p>

<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">일반 매개변수에는 기본값이 없습니다. 호출할 때 값을 전달해야 합니다.</p>
<p style="line-height:1.8;">With a <strong>positional argument</strong> (위치 인수), Python matches values to parameters from left to right.</p>

<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">위치 인수를 사용하면 파이썬은 왼쪽부터 순서대로 매개변수에 값을 연결합니다.</p>
<p style="line-height:1.8;">Here, <code>"Americano"</code> goes to <code>drink</code>. The number <code>2</code> goes to <code>cups</code>.</p>

<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">여기서는 <code>"Americano"</code>가 <code>drink</code>에, 숫자 <code>2</code>가 <code>cups</code>에 전달됩니다.</p>

```python
def order(drink, cups):
    print(cups, "cup(s) of", drink)

order("Americano", 2)
```

<p style="line-height:1.8;">Every regular parameter needs a value. This call is missing <code>cups</code>, so Python raises a <code>TypeError</code>.</p>

<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">일반 매개변수에는 모두 값이 필요합니다. 이 호출에는 <code>cups</code> 값이 빠져 있어 <code>TypeError</code>가 발생합니다.</p>

```python
# This call is incomplete:
# order("Americano")
```

<h2 style="color:#d1242f; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Default parameters</h2>

<p style="line-height:1.8;">A <strong>default parameter</strong> (기본 매개변수) has a value in the function definition.</p>

<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">기본 매개변수는 함수 정의에서 기본값을 지정한 매개변수입니다.</p>
<p style="line-height:1.8;">If the caller leaves it out, Python uses the default. If the caller gives a value, that value replaces the default.</p>

<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">호출할 때 값을 생략하면 기본값을 사용합니다. 값을 전달하면 전달한 값으로 바뀝니다.</p>

<pre><code>def function_name(required, optional=default_value):
    code</code></pre>

<p style="line-height:1.8;">A parameter with no default is required. A parameter with a default is optional.</p>

<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">기본값이 없는 매개변수는 필수이고, 기본값이 있는 매개변수는 선택입니다.</p>

```python
def order(drink, cups=1):
    print(cups, "cup(s) of", drink)

order("Latte")       # uses cups=1
order("Latte", 3)    # uses cups=3
```

<p style="line-height:1.8;">The default is used only when an argument is not provided. It does not stop us from choosing another value.</p>

<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">인수를 전달하지 않았을 때만 기본값을 사용합니다. 다른 값을 선택해 전달할 수 있습니다.</p>
<p style="line-height:1.8;">Parameters without defaults must come before parameters with defaults.</p>

<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">기본값이 없는 매개변수는 기본값이 있는 매개변수보다 앞에 와야 합니다.</p>

<pre><code># Valid
def order(drink, cups=1):
    ...

# Invalid: a required parameter follows a default parameter
# def order(cups=1, drink):
#     ...</code></pre>

<h2 style="color:#8250df; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Keyword arguments</h2>

<p style="line-height:1.8;">A <strong>keyword argument</strong> (키워드 인수) names the parameter when we call a function: <code>parameter=value</code>.</p>

<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">키워드 인수는 함수를 호출할 때 <code>매개변수=값</code> 형식으로 매개변수 이름을 지정합니다.</p>
<p style="line-height:1.8;">This is a way to pass an argument. It is not a different kind of parameter.</p>

<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">키워드 인수는 값을 전달하는 방법입니다. 별도의 매개변수 종류를 뜻하지는 않습니다.</p>
<p style="line-height:1.8;">Keyword arguments make calls easier to read. Their order does not matter because each value names its parameter.</p>

<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">각 값에 매개변수 이름을 붙이므로 순서를 바꿔도 됩니다. 코드를 읽기도 쉽습니다.</p>

```python
def make_sandwich(bread, filling, toasted=False):
    print("Bread:", bread)
    print("Filling:", filling)
    print("Toasted:", toasted)

# Name each argument. We can write them in any order.
make_sandwich(toasted=True, filling="egg", bread="wheat")
```

<p style="line-height:1.8;">We can mix positional and keyword arguments. Put positional arguments first, then keyword arguments.</p>

<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">위치 인수와 키워드 인수를 함께 사용할 수 있습니다. 위치 인수를 먼저 쓰고 키워드 인수를 뒤에 씁니다.</p>
<p style="line-height:1.8;">Here, <code>"sourdough"</code> fills <code>bread</code> by position. The other values are matched by name.</p>

<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">여기서 <code>"sourdough"</code>는 위치로 <code>bread</code>에 전달됩니다. 나머지 값은 이름으로 연결됩니다.</p>

```python
make_sandwich("sourdough", filling="cheese", toasted=True)
```

<p style="line-height:1.8;">Do not give the same parameter a value twice. This call gives <code>bread</code> once by position and again by name, so Python raises a <code>TypeError</code>.</p>

<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">같은 매개변수에 값을 두 번 전달하면 안 됩니다. 이 호출은 <code>bread</code>에 위치와 이름으로 값을 모두 전달해 <code>TypeError</code>가 발생합니다.</p>

<pre><code># Invalid:
# make_sandwich("wheat", bread="rye", filling="egg")</code></pre>

<p style="line-height:1.8;">A required parameter still needs a value. You can provide it by position or by name.</p>

<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">필수 매개변수에도 값이 필요합니다. 위치나 이름을 사용해 전달할 수 있습니다.</p>

<h2 style="color:#1a7f37; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Many positional arguments</h2>

<p style="line-height:1.8;">Use <code>*args</code> when a function should accept any number of extra positional arguments.</p>

<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">함수가 위치 인수를 여러 개 받을 수 있게 하려면 <code>*args</code>를 사용합니다.</p>
<p style="line-height:1.8;">Inside the function, <code>args</code> is a tuple. The name <code>args</code> is a convention; the <code>*</code> does the work.</p>

<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">함수 안에서 <code>args</code>는 튜플입니다. <code>args</code>라는 이름은 관례이고, 실제 역할은 <code>*</code>가 합니다.</p>

```python
def show_items(*items):
    print(items)
    for item in items:
        print("-", item)

show_items("notebook", "pen", "eraser")
```

<h2 style="color:#0550ae; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Keyword-only parameters</h2>

<p style="line-height:1.8;">A parameter after <code>*args</code> must be passed by name. It is called a <strong>keyword-only parameter</strong>.</p>

<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;"><code>*args</code> 뒤에 오는 매개변수는 이름을 지정해서 전달해야 합니다. 이를 키워드 전용 매개변수라고 합니다.</p>
<p style="line-height:1.8;">This is useful when a value could be confused with the many positional arguments.</p>

<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">값이 여러 위치 인수와 혼동될 수 있을 때 유용합니다.</p>

```python
def repeat_items(*items, times=2):
    for _ in range(times):
        for item in items:
            print(item)
        print()

repeat_items("read", "practice", times=3)
```

<h2 style="color:#d1242f; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Key points</h2>

<ul style="line-height:1.9;">
<li><strong>Parameter:</strong> a name in a function definition.<br><span style="font-size:0.85em; opacity:0.7;">매개변수: 함수 정의에 적는 이름입니다.</span></li>
<li><strong>Argument:</strong> a value passed when calling a function.<br><span style="font-size:0.85em; opacity:0.7;">인수: 함수를 호출할 때 전달하는 값입니다.</span></li>
<li><strong>Regular parameter:</strong> needs a value unless it has a default.<br><span style="font-size:0.85em; opacity:0.7;">일반 매개변수: 기본값이 없으면 반드시 값이 필요합니다.</span></li>
<li><strong>Default parameter:</strong> uses its default when no argument is given.<br><span style="font-size:0.85em; opacity:0.7;">기본 매개변수: 인수를 생략하면 기본값을 사용합니다.</span></li>
<li><strong>Positional argument:</strong> matched by its place, from left to right.<br><span style="font-size:0.85em; opacity:0.7;">위치 인수: 왼쪽부터 순서대로 매개변수에 연결됩니다.</span></li>
<li><strong>Keyword argument:</strong> matched by the parameter name, such as <code>cups=2</code>.<br><span style="font-size:0.85em; opacity:0.7;">키워드 인수: <code>cups=2</code>처럼 매개변수 이름으로 연결됩니다.</span></li>
<li>Put required parameters before parameters with defaults.<br><span style="font-size:0.85em; opacity:0.7;">필수 매개변수는 기본값이 있는 매개변수보다 앞에 둡니다.</span></li>
<li>When mixing argument styles, put positional arguments before keyword arguments.<br><span style="font-size:0.85em; opacity:0.7;">두 방식을 함께 쓸 때는 위치 인수를 먼저 씁니다.</span></li>
<li>A keyword-only parameter must be passed by name.<br><span style="font-size:0.85em; opacity:0.7;">키워드 전용 매개변수는 이름을 지정해 전달해야 합니다.</span></li>
</ul>

<p style="line-height:1.8;"><strong>Quick rule:</strong> Position tells Python <em>which parameter</em>. A keyword tells Python by <em>name</em>. A default tells Python what value to use when an argument is missing.<br><span style="font-size:0.85em; opacity:0.7;">위치는 어떤 매개변수인지 알려주고, 키워드는 이름으로 지정합니다. 인수를 생략하면 기본값을 사용합니다.</span></p>

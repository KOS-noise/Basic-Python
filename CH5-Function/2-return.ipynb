<h1 style="color:#0969da; font-size:2.4em; font-weight:700; letter-spacing:-0.02em; margin-bottom:0.35em;">Return values</h1>
<p style="line-height:1.8;">A function can give a value back. This is called a <strong>return value</strong>.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">함수가 돌려주는 값을 <strong>반환값</strong>이라고 합니다.</p>

<p style="line-height:1.8;">A function can return a value. What does that mean? Let’s look at an example.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">함수는 값을 반환할 수 있습니다. 반환이 무엇인지 예제로 살펴봅시다.</p>

<h2 style="color:#8250df; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">End a function with <code>return</code></h2>
<p style="line-height:1.8;">Use <code>return</code> to stop a function. Python skips the code after it.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;"><code>return</code>은 함수를 끝냅니다. 뒤에 있는 코드는 실행하지 않습니다.</p>

```python
def return_test():
    print("A 위치이다.")
    return  # 리턴한다.
    print("B 위치이다.")

return_test()
```

<p style="line-height:1.8;">The first <code>return</code> ends the function, so the second print statement does not run.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">첫 번째 <code>return</code>에서 함수가 끝나므로 두 번째 출력문은 실행되지 않습니다.</p>
<p style="line-height:1.8;"><code>return</code> can also send a value back to the code that called the function. In this example, it ends the function without sending a value.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;"><code>return</code>은 함수를 호출한 곳에 값을 돌려줄 수도 있습니다. 이 예제에서는 값을 돌려주지 않고 함수를 끝냅니다.</p>

<h2 style="color:#1a7f37; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Return a value</h2>

<p style="line-height:1.8;">Write a value after <code>return</code> to send it back. The caller can store the returned value in a variable.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;"><code>return</code> 뒤에 값을 쓰면 그 값을 돌려줍니다. 호출한 곳에서 반환값을 변수에 저장할 수 있습니다.</p>

```python
def return_test1():
    return 100

value = return_test1()

print(value)
```

<h2 style="color:#0550ae; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Return no value</h2>

<p style="line-height:1.8;">What happens if <code>return</code> has no value after it?</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;"><code>return</code> 뒤에 값이 없으면 어떻게 될까요?</p>

```python
def return_none():
    return

value = return_none()
print(value)
```

<p style="line-height:1.8;">The function returns <code>None</code>. Python also returns <code>None</code> if a function ends without a <code>return</code> statement.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">이 함수는 <code>None</code>을 반환합니다. 함수가 <code>return</code> 없이 끝나도 파이썬은 <code>None</code>을 반환합니다.</p>

<h2 style="color:#d1242f; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Use a function result</h2>
<p style="line-height:1.8;">A returned value can be used in an expression, stored in a variable, or printed. The examples below add a range of numbers and return the total.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">반환값은 식에 사용하거나 변수에 저장하거나 출력할 수 있습니다. 아래 예제는 범위 안의 수를 더해 합계를 반환합니다.</p>
<p style="line-height:1.8;">The second function also has default parameters. You can change the range or step when you call it, or use the defaults.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">두 번째 함수에는 기본 매개변수도 있습니다. 호출할 때 범위와 간격을 지정하거나 기본값을 사용할 수 있습니다.</p>

```python
# 범위 내부의 정수를 모두 더하는 함수 

def sum_all(start, end):

    output = 0

    for i in range(start, end + 1):
        output += i

    return output


sum_all(0, 100)
```

```python
# 기본 매개변수와 키워드 매개변수를 활용해 범위의 정수를 더하는 함수 

def sum_all(start=0, end=100, step=1):
    output = 0

    for i in range(start, end + 1, step):
        output += i
    return output
```

```python
print("A.", sum_all(0, 100, 10))
```

```python
print("B.", sum_all(end=100))
```

```python
print("C.", sum_all(end=100, step=2))
```

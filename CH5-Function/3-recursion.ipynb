<h1 style="color:#0969da; font-size:2.4em; font-weight:700; letter-spacing:-0.02em; margin-bottom:0.35em;">Recursive functions</h1>
<p style="line-height:1.8;">A recursive function calls itself to solve a smaller part of a problem.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">재귀 함수는 자기 자신을 호출해 문제를 더 작은 부분으로 나누어 풉니다.</p>

<p style="line-height:1.8;">Do you know factorials?</p>
<pre><code>n! = n × (n - 1) × (n - 2) × ... × 1</code></pre>
<p style="line-height:1.8;">For example, <code>4! = 4 × 3 × 2 × 1 = 24</code>. We can calculate a factorial with a loop or with recursion.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">예를 들어 <code>4! = 4 × 3 × 2 × 1 = 24</code>입니다. 팩토리얼은 반복문이나 재귀로 계산할 수 있습니다.</p>
<ul><li>Use a loop.</li><li>Use a recursive function.</li></ul>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">반복문으로 계산하거나 재귀 함수로 계산할 수 있습니다.</p>

```python
def factorial(n):

    # 변수를 선언합니다.
    output = 1
    # 반복문을 돌려 숫자를 더합니다.
    for i in range(1, n+1):
        output *= i
    # 리턴합니다.
    
    return output
```

```python
print("1!: ", factorial(1))
print("3!: ", factorial(3))
print("80!: ", factorial(80))
```

<h2 style="color:#8250df; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Factorial with recursion</h2>

<p style="line-height:1.8;">Another way is to use recursion. Recursion means that a function calls itself.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">다른 방법은 재귀를 사용하는 것입니다. 재귀는 함수가 자기 자신을 호출하는 방식입니다.</p>
<p style="line-height:1.8;">We can define factorial like this:</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">팩토리얼을 다음 규칙으로 정의할 수 있습니다.</p>

<pre><code>factorial(n) = n × factorial(n - 1)  when n ≥ 1
factorial(0) = 1</code></pre>
<p style="line-height:1.8;">The first rule makes the problem smaller. The second rule is the <strong>base case</strong>. It stops the recursion.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">첫 번째 규칙은 문제를 더 작게 만듭니다. 두 번째 규칙은 재귀를 멈추는 <strong>종료 조건</strong>입니다.</p>

```python
def factorial(n):

    # n 이 0 이라면 1을 return 한다. 
    if n == 0:
        return 1
    else: 
        return n * factorial(n-1)
```

```python
factorial(0)
```

<h2 style="color:#d1242f; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">A problem with recursion</h2>

<p style="line-height:1.8;">A recursive function can be easy to read, but it may repeat the same work many times. For large inputs, it can become slow.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">재귀 함수는 읽기 쉬울 수 있지만 같은 계산을 여러 번 반복할 수 있습니다. 입력이 크면 느려질 수 있습니다.</p>
<p style="line-height:1.8;">First, let’s look at the Fibonacci sequence. Then we can see why its recursive version repeats work. A later technique called <strong>memoization</strong> can save earlier results.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">먼저 피보나치 수열을 살펴보겠습니다. 재귀 함수가 왜 같은 계산을 반복하는지 확인한 뒤, 이전 결과를 저장하는 메모이제이션도 알아봅니다.</p>

<p style="line-height:1.8;">Here is a rabbit population puzzle called the Fibonacci sequence. Under these rules, the numbers are <code>1, 1, 2, 3, 5, 8, 13, ...</code>.</p>
<ul><li>Start with one pair of baby rabbits.</li><li>A pair becomes able to have babies after one month.</li><li>Each adult pair has one new pair every month.</li><li>Rabbits never die.</li></ul>
<p style="line-height:1.8;">The number of pairs this month is the sum of the numbers from the previous two months.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">피보나치 수열을 토끼 개체 수 문제로 설명할 수 있습니다. 한 달의 쌍 수는 직전 두 달의 쌍 수를 더한 값입니다.</p>

<p style="line-height:1.8;"><strong>Fibonacci sequence: how the population grows</strong></p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">피보나치 수열에 따라 토끼 쌍이 늘어나는 모습입니다.</p>
<img src="images/fibonacci-rabbits.png" alt="Cute hand-drawn rabbit pairs connected as a branching Fibonacci family tree." width="600">

<p style="line-height:1.8;">The sequence grows like this: <code>1, 1, 2, 3, 5, 8, 13, ...</code>. Let’s describe the rule.</p>
<ul><li>The first value is 1.</li><li>The second value is 1.</li><li>Each later value is the sum of the two values before it.</li></ul>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">수열은 <code>1, 1, 2, 3, 5, 8, 13, ...</code>처럼 자랍니다. 첫 두 값은 1이고, 그다음부터는 앞의 두 값을 더합니다.</p>

```python
def fibonacci(n):
    if n == 1:
        return 1
    elif n == 2:
        return 1
    else:
        return fibonacci(n-1) + fibonacci(n-2)
```

```python
fibonacci(4)
```

<p style="line-height:1.8;">This simple function works, but try calling <code>fibonacci(35)</code>. It may take a few seconds.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">이 함수는 간단하지만 <code>fibonacci(35)</code>를 호출하면 몇 초가 걸릴 수 있습니다.</p>

```python
fibonacci(35)
```

<p style="line-height:1.8;">Why is it slow? Each call may start more calls, and many of them calculate the same values again.</p>
<p style="line-height:1.8;">We can draw these calls as a <strong>tree</strong>. Each box is one function call. A branch shows the calls it makes.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">각 호출이 다시 여러 함수를 부르고, 같은 값을 반복해서 계산하기 때문에 느립니다. 호출을 트리로 그리면 각 상자는 함수 호출이고 가지는 그 함수가 부르는 호출입니다.</p>

<p style="line-height:1.8;"><strong>Call tree for <code>fibonacci(5)</code></strong></p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;"><code>fibonacci(5)</code>의 재귀 호출 트리입니다. 같은 계산이 여러 번 나타나는 것을 확인해 보세요.</p>
<img src="images/fibonacci-call-tree.svg" alt="A tree diagram of recursive calls made by fibonacci(5), showing repeated calls to fibonacci(3), fibonacci(2), and fibonacci(1)." width="760">

<p style="line-height:1.8;">To find a value at the bottom of the tree, the function must follow each branch. Some calls appear again in other branches.</p>
<p style="line-height:1.8;">Because of this repeated work, the number of calls grows quickly. Saving results can make the function much faster. This is called <strong>memoization</strong>.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">트리 아래쪽의 값을 구하려면 각 가지의 호출을 따라가야 합니다. 여러 가지에서 같은 호출이 반복되므로 전체 호출 횟수가 빠르게 늘어납니다. 계산 결과를 저장해 두는 방법을 메모이제이션이라고 합니다.</p>

```python
counter = 0

def fibonacci(n):
    print(f"fibonacci({n})를 구한다.")

    global counter  # 여기 주목
    counter += 1

    if n == 1:
        return 1
    elif n == 2:
        return 1
    else:
        return fibonacci(n-1) + fibonacci(n-2)
```

```python
fibonacci(10)
```

<p style="line-height:1.8;">A function cannot directly change a global variable just by reading it. If we assign to that name inside a function, Python treats it as a local variable. This can cause an <code>UnboundLocalError</code>.</p>
<p style="line-height:1.8;">The <code>global</code> keyword tells Python to use the variable outside the function. The next example uses it to count function calls.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">함수 안에서 전역 변수 이름에 값을 대입하면 파이썬은 이를 지역 변수로 취급해 <code>UnboundLocalError</code>가 발생할 수 있습니다. <code>global</code> 키워드는 함수 바깥의 변수를 사용한다고 알려줍니다. 다음 예제는 함수 호출 횟수를 셉니다.</p>

<h2 style="color:#1a7f37; font-size:1.7em; font-weight:700; margin-bottom:0.4em;"><code>global</code> keyword</h2>
<p style="line-height:1.8;">Use <code>global</code> only when you need to assign to a global variable inside a function. Often, returning a value is a clearer choice.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">함수 안에서 전역 변수에 값을 대입해야 할 때 <code>global</code>을 사용합니다. 보통은 값을 반환하는 방식이 더 명확합니다.</p>

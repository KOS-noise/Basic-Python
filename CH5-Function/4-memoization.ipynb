<h1 style="color:#0969da; font-size:2.4em; font-weight:700; letter-spacing:-0.02em; margin-bottom:0.35em;">Memoization</h1>
<p style="line-height:1.8;">Memoization saves a function’s results so it can reuse them later.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">메모이제이션은 함수의 계산 결과를 저장해 두었다가 다시 사용하는 방법입니다.</p>

<p style="line-height:1.8;">We saw that the recursive Fibonacci function can be slow. It repeats the same calculations many times.</p>
<p style="line-height:1.8;">We can save each result and reuse it. This makes the function much faster.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">재귀 피보나치 함수는 같은 계산을 여러 번 반복해서 느릴 수 있습니다. 계산 결과를 저장해 두면 다시 계산하지 않아도 되어 더 빨라집니다.</p>

```python
# 메모 변수를 만든다.
dictionary = {
    1: 1,
    2: 1
}
```

```python
# 함수를 선언한다.
def fibonacci(n):
    if n in dictionary:
        # 메모가 되어 있으면 메모된 값을 리턴
        return dictionary[n]
    else:
        #메모가 되어 있지 않으면 값을 구함
        output = fibonacci(n-1) + fibonacci(n-2)
        dictionary[n] = output
        return output
```

```python
print("fibonacci(10): ", fibonacci(10))
print("fibonacci(20): ", fibonacci(20))
print("fibonacci(30): ", fibonacci(30))
print("fibonacci(40): ", fibonacci(40))
print("fibonacci(50): ", fibonacci(50))
```

<p style="line-height:1.8;">A dictionary can store the values we have already calculated. This is called <strong>memoization</strong>.</p>
<p style="line-height:1.8;">When a value is in the dictionary, the function returns it right away. Otherwise, it calculates the value, saves it, and returns it.</p>
<p style="line-height:1.8;">Without memoization, <code>fibonacci(50)</code> can take a long time. With memoization, it finishes much faster.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">계산한 값을 딕셔너리에 저장하는 것을 메모이제이션이라고 합니다. 값이 이미 있으면 바로 반환하고, 없으면 계산한 뒤 저장합니다. 메모이제이션을 사용하면 <code>fibonacci(50)</code>도 훨씬 빠르게 계산할 수 있습니다.</p>

<p style="line-height:1.8;">We can add all items from a list with <code>+</code> or <code>extend()</code>. Here, we use the <code>+</code> operator.</p>

<p style="line-height:1.8;">The result is still nested: <code>[5, 6]</code> remains inside a list.</p>
<p style="line-height:1.8;">To flatten every level, call <code>flatten()</code> again when an item is a list. Add the returned flat values to <code>output</code>.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">결과에 <code>[5, 6]</code>이 아직 리스트로 남아 있습니다. 모든 단계의 중첩을 없애려면 원소가 리스트일 때 <code>flatten()</code>을 다시 호출하고, 반환된 값을 <code>output</code>에 추가합니다.</p>

```python
def flatten(data):
    output = []

    for item in data:
        if type(item) == list:
            output += flatten(item)

        else:
            output.append(item)

    return output
```

```python
print("원본: ", example)
print("변환: ", flatten(example))
```

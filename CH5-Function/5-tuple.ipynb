<p style="line-height:1.8;">Python has useful features that make functions more flexible. This section covers tuples and lambda expressions.</p>
<ul><li><strong>Tuple:</strong> like a list, but you cannot change its items.</li><li><strong>Lambda:</strong> a short function written in one expression. It is useful for simple tasks.</li></ul>

<h1 style="color:#0969da; font-size:2.4em; font-weight:700; letter-spacing:-0.02em; margin-bottom:0.35em;">Tuple</h1>

<p style="line-height:1.8;">A tuple is similar to a list. The main difference is that a tuple cannot be changed after it is created.</p>
<p style="line-height:1.8;">Write its items inside parentheses:</p>
<pre><code>(item, item, item, ...)</code></pre>

```python
tuple_test = (10, 20, 30)

tuple_test[0]
```

```python
tuple_test[1]
```

<p style="line-height:1.8;">We can read tuple items just like list items. But we cannot change an item.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">튜플은 항목을 바꿀 수 없는 자료형입니다. 이를 불변(immutable)이라고 합니다.</p>

```python
tuple_test[0] = 1
```

<h2 style="color:#8250df; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">A tuple with one item</h2>
<p style="line-height:1.8;">A one-item list is easy to write. A one-item tuple needs a comma.</p>

```python
[273]
```

<p style="line-height:1.8;">How do we make a tuple with one item?</p>
<p style="line-height:1.8;"><code>(273)</code> is just the number <code>273</code> in parentheses. Add a comma to make a one-item tuple: <code>(273,)</code>.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">원소가 하나인 튜플은 쉼표가 꼭 필요합니다. <code>(273)</code>은 정수이고, <code>(273,)</code>은 튜플입니다.</p>

```python
(273, )
```

<h2 style="color:#1a7f37; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Parentheses are optional</h2>

```python
[a, b] = [10, 20]
(c, d) = (10, 20)
```

```python
print(a, b, c, d)
```

<p style="line-height:1.8;">Lists and tuples can both assign items to several variables.</p>
<p style="line-height:1.8;">For a tuple, parentheses are optional. Python still treats comma-separated values as a tuple.</p>

```python
tuple_test = 10, 20, 30, 40

print("# 괄호가 없는 튜플의 값과 자료형 출력")
print("tuple_test:", tuple_test)
print(type(tuple_test))
```

<p style="line-height:1.8;">The output is a tuple: <code>(10, 20, 30, 40)</code>.</p>

```python
# 괄호가 없는 튜플 활용

a, b, c = 10, 20, 30

print("# 괄호가 없는 튜플을 활용한 할당")
print(a, b, c)
```

<h2 style="color:#0550ae; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Swap values with a tuple</h2>

```python
a, b = 10, 20

print("# 교환 전 값")
print("a: ", a)
print("b: ", b)
print()

# 값을 교환한다.
a, b = b, a

print("# 교환 후 값")
print("a: ", a)
print("b: ", b)
```

<p style="line-height:1.8;">The line <code>a, b = b, a</code> swaps the values. Python packs the values into a tuple, then assigns them in the new order.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">튜플을 이용하면 임시 변수 없이 두 변수의 값을 바꿀 수 있습니다.</p>

<h2 style="color:#d1242f; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Tuples and functions</h2>

<p style="line-height:1.8;">A function can return a tuple. We can assign each returned item to a variable in one line.</p>

```python
def test():
    return(10, 20)

# 여러 개의 값을 리턴받기
a, b = test()

print("a: ", a)
print("b: ", b)
```

<p style="line-height:1.8;">Tuples support operators like <code>+</code> and <code>*</code>, just like lists.</p>
<p style="line-height:1.8;">One key difference is that tuples cannot be changed. Tuple unpacking also lets us assign several values at once.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">핵심은 튜플의 항목은 수정할 수 없다는 점입니다. 쉼표로 나열한 값을 여러 변수에 한 번에 할당할 수도 있습니다.</p>

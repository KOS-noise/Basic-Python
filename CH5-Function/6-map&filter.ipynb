<p style="line-height:1.8;">In Python, we can pass a function to another function. Built-in functions like <code>map()</code> and <code>filter()</code> use this idea.</p>

<h1 style="color:#0969da; font-size:2.4em; font-weight:700; letter-spacing:-0.02em; margin-bottom:0.35em;">Pass a function as an argument</h1>

<p style="line-height:1.8;">A function can receive another function as an argument. A function passed this way is often called a <strong>callback function</strong>.</p>
<p style="line-height:1.8;">In the next example, <code>call_10_times</code> receives a function and calls it inside its loop. Notice that we pass <code>print_hello</code> without parentheses.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">다른 함수에 전달한 함수를 콜백 함수라고 부릅니다. 괄호 없이 함수 이름을 전달하면 함수 자체가 전달됩니다.</p>

```python
def call_10_times(func):
    for i in range(5):
        func()

# 간단한 출력하는 함수
def print_hello():
    print("안녕하세요")


call_10_times(print_hello)
```

<h2 style="color:#8250df; font-size:1.7em; font-weight:700; margin-bottom:0.4em;"><code>map()</code> and <code>filter()</code></h2>

<p style="line-height:1.8;"><code>map()</code> and <code>filter()</code> are built-in functions. Both take a function and an iterable, such as a list.</p>
<p style="line-height:1.8;"><code>map()</code> applies the function to each item. It gives us the results one by one.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;"><code>map()</code>은 각 원소를 함수에 넣어 변환하고, <code>filter()</code>는 조건에 맞는 원소를 고릅니다.</p>

```python
map(함수, 리스트)
```

<p style="line-height:1.8;"><code>filter()</code> checks each item with a function. It keeps an item only when the function returns <code>True</code>.</p>

```python
filter(함수, 리스트)
```

```python
# 함수 선언
def power(item):
    return item
def under_3(item):
    return item < 3


# 변수 선언
list_input_a = [1, 2, 3, 4, 5]
```

```python
# map() 함수 사용
output_a = map(power, list_input_a)
print("# map() 함수의 실행 결과")
print("map(power, list_input_a): ", output_a)
print("map(power, list_input_a): ", list(output_a))
print()
```

```python
# filter() 함수 사용
output_b = filter(under_3, list_input_a)
print("# filter() 함수의 실행 결과")
print("filter(under_3, list_input_a): ", output_b)
print("filter(under_3, list_input_a): ", list(output_b))
```

<p style="line-height:1.8;">Both functions take a function first and an iterable second.</p>
<p style="line-height:1.8;">They return iterator objects: a <code>map</code> object or a <code>filter</code> object. These are not lists. They produce items as we use them.</p>
<p style="line-height:1.8;">Use <code>list()</code> to collect the items into a list. This also uses up the iterator.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;"><code>map()</code>과 <code>filter()</code>는 리스트가 아니라 반복자(iterator)를 반환합니다. <code>list()</code>로 리스트로 바꿀 수 있지만, 반복자의 값을 모두 사용합니다.</p>

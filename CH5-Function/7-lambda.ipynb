<h1 style="color:#0969da; font-size:2.4em; font-weight:700; letter-spacing:-0.02em; margin-bottom:0.35em;">Lambda expressions</h1>

<p style="line-height:1.8;">Writing a full function with <code>def</code> can feel long when the function is very simple. A <strong>lambda expression</strong> lets us write a short function in one line.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">람다 표현식은 간단한 함수를 한 줄로 만들 때 사용합니다.</p>

```python
lambda 매개변수: 리턴값
```

<p style="line-height:1.8;">Let’s rewrite the <code>power()</code> and <code>under_3()</code> functions from the previous example as lambda expressions.</p>
<p style="line-height:1.8;">A lambda has parameters, a colon, and one expression. It returns the result of that expression automatically. We do not write <code>return</code>.</p>
<pre><code>lambda parameters: expression</code></pre>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">람다는 한 개의 표현식 결과를 자동으로 반환합니다. 함수 본문은 한 줄의 표현식이어야 합니다.</p>

```python
# declare function
power = lambda x: x * x
under_3 = lambda x: x < 3
```

```python
# declare value
list_input_a = [1, 2, 3, 4, 5]
```

```python
# map() function
output_a = map(power, list_input_a)
print("# map() 함수의 실행 결과")
print("map(power, list_input_a): ", output_a)
print("map(power, list_input_a): ", list(output_a))
```

```python
# filter() function
output_b = filter(under_3, list_input_a)
print("# filter() 함수의 실행 결과")
print("map(under_3, list_input_a): ", output_b)
print("map(under_3, list_input_a): ", list(output_b))
```

<p style="line-height:1.8;">The lambda expressions above do the same work as the named functions. Let’s use them directly inside <code>map()</code> and <code>filter()</code>.</p>
<p style="line-height:1.8;">This is useful when a short function is needed in only one place.</p>

```python
# inline lambda

list_input_a = [1, 2, 3, 4, 5]

output_a = map(lambda x: x * x, list_input_a) # power() 함수로 선언하지도 않고, 매개변수로 바로 넣었음 
print("# map() 함수의 실행 결과")
print("map(power, list_input_a): ", output_a)
print("map(power, list_input_a): ", list(output_a))
print()

output_b = filter(lambda x: x < 3, list_input_a) # under_3() 함수를 선언하지도 않고 매개변수로 바로 넣었음 
print("# filter() 함수의 실행 결과")
print("map(under_3, list_input_a): ", output_b)
print("map(under_3, list_input_a): ", list(output_b))
```

<p style="line-height:1.8;">The result is the same, but the code is shorter with lambda expressions.</p>
<p style="line-height:1.8;">Use a named function when you want to reuse it or give it a clear name. Use a lambda for a short function you need only once.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">여러 곳에서 재사용하거나 이름이 필요한 함수는 <code>def</code>로 정의하고, 한 번만 쓸 간단한 함수는 람다로 쓸 수 있습니다.</p>

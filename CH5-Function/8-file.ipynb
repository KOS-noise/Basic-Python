<h1 style="color:#0969da; font-size:2.4em; font-weight:700; letter-spacing:-0.02em; margin-bottom:0.35em;">File handling</h1>

<p style="line-height:1.8;">Python has built-in functions for working with files. We can use them with text files and binary files.</p>
<p style="line-height:1.8;">To work with a file, open it first. Choose whether to read from it or write to it.</p>

```python
# 파일을 열 때는 open() 함수를 사용한다.

파일 객체 = open(문자열: 파일 경로, 문자열: 읽기 모드)
```

<p style="line-height:1.8;"><code>open()</code> takes a file path and a mode. The mode tells Python what you want to do.</p>

| Mode | Meaning |
|---|---|
| <code>"w"</code> | Write. Creates a file or replaces its contents. |
| <code>"a"</code> | Append. Adds text to the end of a file. |
| <code>"r"</code> | Read. Opens a file to read it. |

```python
# 파일을 닫을 때는 close() 함수를 사용합니다.

파일 객체.close()
```

```python
# example code

file = open("basic.txt", "w")

# write some text on the file

file.write("Hello! I'm KOS noise. I want to be a good programmer.")
```

<p style="line-height:1.8;">After you run this cell, <code>basic.txt</code> appears in the notebook’s current folder. It contains the text written by the code.</p>

```python
# close file
file.close()
```

<p style="line-height:1.8;">Normally, close a file with <code>close()</code> when you finish.</p>
<p style="line-height:1.8;">In a notebook, the file may stay open until you close it. A safer and easier way is to use <code>with</code>.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">파일을 닫지 않으면 변경 내용이 저장되지 않거나 다른 작업에 영향을 줄 수 있습니다. <code>with</code>를 쓰면 블록이 끝날 때 파일이 자동으로 닫힙니다.</p>

<h2 style="color:#8250df; font-size:1.7em; font-weight:700; margin-bottom:0.4em;"><code>with</code> statement</h2>

<p style="line-height:1.8;">Without <code>with</code>, we must remember to close the file. If an error happens first, we might forget.</p>
<p style="line-height:1.8;">The <code>with</code> statement closes the file automatically when its indented block ends.</p>

```python
with open(문자열: 파일 경로, 문자열: 모드) as 파일 객체:
    문장
```

<p style="line-height:1.8;">We can rewrite the earlier example with <code>with</code>. The file closes automatically after the block.</p>

```python
# 파일을 엽니다
with open("basic.txt", "w") as file:
    # 파일에 텍스트를 씁니다.
    file.write("Hello Python?")
```

<p style="line-height:1.8;">With <code>with</code>, the file closes when the indented block ends.</p>

<h2 style="color:#1a7f37; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Streams</h2>
<p style="line-height:1.8;">Data moves between a program and another place, such as a file or a network. This flow is called a <strong>stream</strong>.</p>
<p style="line-height:1.8;"><code>open()</code> opens a stream between the program and a file. <code>close()</code> ends that stream.</p>

<h2 style="color:#0550ae; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Read text</h2>
<p style="line-height:1.8;">Use the file object’s <code>read()</code> method to read its contents.</p>

```python
파일 객체.read()
```

```python
with open("basic.txt", "r") as file:
    contents = file.read()

print(contents)
```

<h2 style="color:#d1242f; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Read text line by line</h2>

<p style="line-height:1.8;">Common text formats for storing data include CSV, XML, and JSON.</p>
<p style="line-height:1.8;"><strong>CSV</strong> means <em>Comma-Separated Values</em>. It stores values separated by commas.</p>

<p style="line-height:1.8;"><strong>Example: CSV data</strong></p>

<pre><code>Name, Height, Weight
Ahn Gildong, 176, 62
Kim Taeyoung, 169, 50</code></pre>

<p style="line-height:1.8;">Each CSV line is one row of data. Commas separate the values in that row.</p>
<p style="line-height:1.8;">The first row is often a <strong>header</strong>. It names the columns and explains what each value means.</p>
<p style="line-height:1.8;">To process a large file, read it one line at a time. This avoids loading the whole file into memory.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">CSV는 행과 열로 데이터를 나타냅니다. 헤더는 각 열의 의미를 알려줍니다. 큰 파일은 한 줄씩 읽으면 메모리를 아낄 수 있습니다.</p>

<p style="line-height:1.8;">The next code creates <code>info.txt</code> with 1,000 rows. Each row has a name, weight, and height.</p>

```python
import random

# 간단한 한글 리스트 만들기
hanguls = list("가나다라마바사아자차카타파하")

# 파일을 쓰기 모드로 엽니다.
with open("info.txt", "w") as file:
    for i in range(1000):
        # 랜덤한 값으로 변수 생성
        name = random.choice(hanguls) + random.choice(hanguls)
        weight = random.randrange(40, 100)
        height = random.randrange(130, 200)

        # 텍스트를 씁니다. 
        file.write("{}, {}, {}\n".format(name, weight, height))
```

<p style="line-height:1.8;"><code>info.txt</code> now contains 1,000 rows of sample data.</p>
<p style="line-height:1.8;">We can read one line at a time with a <code>for</code> loop.</p>

```python
for 한 줄을 나타내는 문자열 in 파일 객체:
    처리
```

<p style="line-height:1.8;">Now read each row from the file and calculate that person’s BMI from their weight and height.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">BMI(체질량지수)는 몸무게(kg)를 키(m)의 제곱으로 나누어 계산합니다.</p>

```python
with open("info.txt", "r") as file:

    for line in file:

        # 변수 선언
        (name, weight, height) = line.strip().split(", ")

        # 데이터에 문제가 없는지 확인
        if (not name) or (not weight) or (not height):
            continue

        # 결과를 계산
        bmi = int(weight) / ((int(height) / 100) ** 2)
        result = " "

        if 25 < bmi:
            result = "과체중"
        elif 18.5 <= bmi:
            result = "정상 체중"
        else:
            result = "저체중"

    
        print('\n'.join([
            f"이름: {name}",
            f"몸무게: {weight}",
            f"키: {height}",
            f"결과: {result}",
        ]))
        print()
```

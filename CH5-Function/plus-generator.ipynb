<h1 style="color:#0969da; font-size:2.4em; font-weight:700; letter-spacing:-0.02em; margin-bottom:0.35em;">Generators</h1>

<p style="line-height:1.8;">A <strong>generator</strong> produces values one at a time, when we ask for them.</p>
<p style="line-height:1.8;">A generator function uses <code>yield</code>. When it reaches <code>yield</code>, it gives back a value and pauses. It remembers where it stopped.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">제너레이터는 값을 하나씩 만들고, <code>yield</code>에서 잠시 멈춥니다. 다음 값을 요청하면 멈춘 곳부터 이어서 실행합니다.</p>

<h2 style="color:#8250df; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Why use a generator?</h2>
<p style="line-height:1.8;">A list keeps all its values in memory. A generator makes the next value only when needed, so it does not need to keep the whole sequence.</p>
<p style="line-height:1.8;">This is useful for large data, file lines, or a sequence that may never end. This step-by-step approach is called <strong>lazy evaluation</strong>.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">큰 데이터를 한꺼번에 리스트로 만들지 않고 필요한 값만 차례로 처리하면 메모리를 아낄 수 있습니다.</p>

```python
# 제너레이터 함수

# 함수 선언
def test():
    print("call function")
    yield "test"

# 함수 호출
print("A spot passed")
test()

print("B spot passed")
test()
print(test())
```

<p style="line-height:1.8;">Calling a generator function does not run its body. It creates a <strong>generator object</strong>.</p>
<p style="line-height:1.8;">In this example, <code>test()</code> is called twice. Both calls create generator objects, but neither one reaches <code>yield</code>. So <code>"call function"</code> is not printed. The last line prints the generator object itself.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">함수를 호출해도 본문은 바로 실행되지 않습니다. 제너레이터 객체만 만들어지며, <code>next()</code>로 값을 요청해야 실행이 시작됩니다.</p>

<p style="line-height:1.8;">Use <code>next()</code> to ask the generator for one value. It runs until the next <code>yield</code>, returns that value, and pauses again.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;"><code>next()</code>를 호출할 때마다 다음 <code>yield</code>까지 실행하고, 그 값을 돌려준 뒤 멈춥니다.</p>

<h2 style="color:#1a7f37; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Follow each <code>next()</code> call</h2>
<ol style="line-height:1.9;">
<li><code>output = test()</code> creates a generator. The function body has not run yet.</li>
<li>The first <code>next(output)</code> prints <code>A spot passed</code>, then yields <code>1</code>.</li>
<li>The second call resumes after the first <code>yield</code>, prints <code>B spot passed</code>, then yields <code>2</code>.</li>
<li>The third call resumes, prints <code>C spot passed</code>, and reaches the end. There is no next value, so Python raises <code>StopIteration</code>.</li>
</ol>
<p style="line-height:1.8;">Because the third call raises an error, <code>print(c)</code> is not reached. This cell shows what happens when we ask for one value too many.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;"><code>StopIteration</code>은 제너레이터가 끝나 더 돌려줄 값이 없다는 뜻입니다.</p>

```python
# 함수 선언

def test():
    print("A spot passed")
    yield 1
    print("B spot passed")
    yield 2
    print("C spot passed")

# 함수 호출
output = test()

# next() 함수를 호출합니다.
print("D spot passed")
a = next(output)
print(a)

print("E spot passed")
b = next(output)
print(b)

print("F spot passed")
c = next(output)
print(c)

next(output)
```

<h2 style="color:#d1242f; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Key points</h2>
<ul style="line-height:1.9;">
<li><code>yield</code> returns one value and pauses the function.</li>
<li><code>next()</code> resumes it and asks for the next value.</li>
<li>A generator remembers its place between calls.</li>
<li>When it has no more values, it raises <code>StopIteration</code>.</li>
<li>A <code>for</code> loop handles that ending automatically.</li>
</ul>
<p style="line-height:1.8;"><strong>In short:</strong> a generator is like a dispenser. It gives one value when asked, then waits for the next request.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">제너레이터는 요청할 때마다 값 하나를 내주는 장치처럼 생각하면 됩니다.</p>

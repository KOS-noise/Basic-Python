<p style="line-height:1.8;">We can use <code>min()</code> and <code>max()</code> to find the smallest and largest values in a list.</p>

```python
a = [52, 273, 32, 103, 57]

min(a)
```

<p style="line-height:1.8;">What if each item is a dictionary, and we want the cheapest or most expensive book?</p>
<p style="line-height:1.8;">Use the <code>key</code> keyword argument. It tells <code>min()</code> or <code>max()</code> which value to compare for each item.</p>
<p style="line-height:1.8;">Here, the key function gets each book’s price. Then <code>min()</code> and <code>max()</code> choose the book with the lowest or highest price.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;"><code>key</code>에 전달한 함수가 각 원소에서 비교할 값을 꺼냅니다. 함수는 비교 기준을 돌려주고, <code>min()</code>이나 <code>max()</code>는 해당 원소를 선택합니다.</p>

```python
books = [{
    "title" : "river boy",
    "price" : 18000
},{
    "title" : "the giver",
    "price" : 26000
},{
    "title" : "the secret garden",
    "price" : 24000
}]
```

```python
def 가격추출함수(book):
    return book["price"]  # 딕셔너리에서 '가격' 값을 추출하는 함수를 선언한다.
```

```python
print("# most cheap book")
print(min(books, key=가격추출함수))  # '가격' 값을 비교하여 최솟값을 구한다.
print()

print("# most expensive book")
print(max(books, key=가격추출함수))  
print()
```

<p style="line-height:1.8;">We can replace the named key function with a lambda expression. Let’s rewrite the example.</p>

```python
print("# most cheap book")
print(min(books, key = lambda book: book['price']))  # '가격' 값을 비교하여 최솟값을 구한다.
print()

print("# most expensive book")
print(max(books, key = lambda x: x['price']))  # '가격' 값을 비교하여 최솟값을 구한다.
print()
```

<p style="line-height:1.8;">Use the list method <code>sort()</code> to sort list items.</p>
<p style="line-height:1.8;"><code>sort()</code> changes the original list. It does not create a new sorted list.</p>

```python
#  sort 함수로 오름차순 정렬하기
a = [52, 273, 103, 32, 57, 272]
a.sort()
a
```

<p style="line-height:1.8;">By default, <code>sort()</code> sorts in ascending order. Set <code>reverse=True</code> to sort in descending order.</p>

```python
a.sort(reverse=True)
a
```

<p style="line-height:1.8;">What if we want to sort dictionaries by one of their values, such as each book’s price?</p>
<p style="line-height:1.8;">Use the <code>key</code> keyword argument with <code>sort()</code>. Give it a function that returns the value to sort by.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">딕셔너리 자체가 아니라 가격처럼 특정 값으로 정렬하려면 <code>key</code> 함수가 그 기준값을 반환하도록 합니다.</p>

```python
print(" 가격 오름차순 정렬 ")
books.sort(key = lambda book: book['price'])
for book in books:
    print(book)
```

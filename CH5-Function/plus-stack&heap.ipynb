<h1 style="color:#0969da; font-size:2.4em; font-weight:700; letter-spacing:-0.02em; margin-bottom:0.35em;">Stack and heap</h1>
<p style="line-height:1.8;">A program needs memory to keep names and values while it runs. This lesson uses a simple model: the <strong>stack</strong> keeps local names and references, and the <strong>heap</strong> holds objects such as lists and dictionaries.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">스택은 함수의 이름과 참조를, 힙은 리스트·딕셔너리 같은 객체를 담는다고 단순하게 생각해 봅시다.</p>

<h2 style="color:#8250df; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">A simple memory picture</h2>
<p style="line-height:1.8;">The stack is organized into frames. A function call adds a frame. When the function returns, that frame is removed.</p>
<p style="line-height:1.8;">The heap holds objects that can vary in size. A variable can store a <strong>reference</strong> (a way to find an object in memory).</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">참조(reference)는 메모리에 있는 객체를 가리키는 연결 정보입니다.</p>

<p style="line-height:1.8;"><strong>Names in the stack model, objects in the heap</strong></p>
<img src="images/stack-heap-memory.svg" alt="A simplified diagram: variables a, b, and c hold simple values; d and e hold references that point to a list and a dictionary in the heap." width="760">

<p style="line-height:1.8;">This drawing is a beginner-friendly model. In Python, <strong>all values are objects</strong>, including numbers and strings. Variable names refer to objects. The exact memory layout depends on the Python implementation.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">그림은 개념을 위한 단순화입니다. 실제 파이썬에서는 숫자와 문자열도 모두 객체입니다.</p>

```python
a = 10
b = True
c = "Hello"
d = [1, 2, 3, 4, 5]
e = {"name": "Gureum", "age": 6}
```

<p style="line-height:1.8;">Here, <code>a</code>, <code>b</code>, and <code>c</code> are names bound to values. <code>d</code> refers to a list object, and <code>e</code> refers to a dictionary object.</p>
<p style="line-height:1.8;">The list and dictionary can change size, so the diagram places them in the heap. The names <code>d</code> and <code>e</code> let the program find them.</p>

<h2 style="color:#1a7f37; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">A function gets its own frame</h2>
<p style="line-height:1.8;">When a function runs, its parameters and local names belong to that call. They are separate names from variables outside the function, even when they refer to the same object.</p>

```python
def greet(name):
    message = "Hello, " + name
    return message

person = "Mina"
greet(person)
```

<p style="line-height:1.8;">During the call, <code>name</code> and <code>message</code> are local names in the function frame. After the function returns, that frame is removed. The returned string can still be used by the caller.</p>
<p style="line-height:1.8;">Python manages memory for us. When an object is no longer reachable, Python can reclaim it.</p>

<h2 style="color:#8250df; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Function frames and variable scope</h2>
<p style="line-height:1.8;">A function call creates a new <strong>local frame</strong>. Its parameters and local variables live there. The caller’s variables stay in the outer, or <strong>global</strong>, frame.</p>
<p style="line-height:1.8;">The argument value is assigned to a new parameter name. The names are separate, even if they refer to the same object.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">함수의 매개변수와 지역 변수는 함수 프레임에 생깁니다. 바깥 변수와는 다른 이름입니다.</p>

<p style="line-height:1.8;"><strong>While the function is running</strong></p>
<img src="images/function-stack-frames.svg" alt="A global frame has a variable value. A function call adds a separate local frame with a parameter and local variable; the local frame is removed when the function returns." width="760">

```python
def add_tax(price):
    tax = 10
    return price + tax

price = 100
add_tax(price)
```

<p style="line-height:1.8;">During the call, <code>price</code> in the function is a parameter, and <code>tax</code> is a local variable. They belong to the function call. When the function returns, its local frame is removed.</p>

<h2 style="color:#1a7f37; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Local names and global names</h2>
<p style="line-height:1.8;">A function can read a global name. But if we assign to that name inside the function, Python treats it as local unless we declare it with <code>global</code>.</p>
<p style="line-height:1.8;">This code raises <code>UnboundLocalError</code> because Python sees <code>a = 20</code> and treats <code>a</code> as local throughout the function. The earlier <code>print(a)</code> tries to read that local name before it has a value.</p>
<pre><code>a = 10

def change():
    print(a)
    a = 20</code></pre>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">함수 안에서 이름에 대입하면 기본적으로 지역 변수로 취급됩니다. 대입 전에 읽으면 <code>UnboundLocalError</code>가 날 수 있습니다.</p>

<p style="line-height:1.8;">Use <code>global</code> only when you really need to reassign a global variable from inside a function.</p>
<pre><code>count = 10

def change_count():
    global count
    count = 20</code></pre>
<p style="line-height:1.8;">Often, it is clearer to return a new value instead of changing a global variable.</p>

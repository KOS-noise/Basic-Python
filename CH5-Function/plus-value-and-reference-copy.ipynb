<h1 style="color:#0969da; font-size:2.4em; font-weight:700; letter-spacing:-0.02em; margin-bottom:0.35em;">Values and references in function calls</h1>
<p style="line-height:1.8;">When we pass an argument to a function, the parameter becomes a new name for the same object. What happens next depends on whether the object can be changed.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">함수의 매개변수는 전달된 객체를 가리키는 새 이름입니다. 객체를 바꿀 수 있는지에 따라 결과가 달라집니다.</p>

<h2 style="color:#8250df; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">First: names and objects</h2>
<p style="line-height:1.8;">Python passes an object reference to the parameter. It does not automatically make a copy of the object.</p>
<p style="line-height:1.8;">For a simple value like an integer, assigning a new value to the parameter does not change the caller’s variable. This is often called <strong>value-copy behavior</strong> in beginner lessons.</p>
<p style="line-height:1.8;">For a mutable object like a list, both names can refer to the same object. Changing that object inside the function is visible to the caller. This is often called <strong>reference-copy behavior</strong>.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">정확히는 값이나 리스트를 자동 복사하지 않습니다. 같은 객체를 가리키는 이름이 함수 안에 하나 더 생깁니다.</p>

<p style="line-height:1.8;"><strong>Rebinding a number vs. changing a shared list</strong></p>
<img src="images/function-value-reference.svg" alt="Two panels compare a local integer name being rebound without changing the caller's name, and two names referring to the same list whose contents change." width="760">

<h2 style="color:#1a7f37; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Example 1: an integer</h2>
<p style="line-height:1.8;">The function gives its local parameter a new value. The caller’s variable still refers to <code>10</code>.</p>

```python
def change_number(number):
    number = 20

value = 10
change_number(value)
print(value)
```

<p style="line-height:1.8;">The output is <code>10</code>. The line <code>number = 20</code> changes which object the local name <code>number</code> refers to. It does not change <code>value</code>.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">함수 안에서 매개변수 이름을 새 값에 연결해도 바깥 변수의 연결은 바뀌지 않습니다.</p>

<h2 style="color:#0550ae; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Example 2: a list</h2>
<p style="line-height:1.8;">A list can be changed. The parameter and the caller’s variable refer to the same list, so an edit to the list is visible outside the function.</p>

```python
def add_item(items):
    items.append(3)

numbers = [1, 2]
add_item(numbers)
print(numbers)
```

<p style="line-height:1.8;">The output is <code>[1, 2, 3]</code>. The function changed the shared list object.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">리스트 자체를 수정하면 함수 밖에서도 수정된 내용이 보입니다.</p>

<h2 style="color:#d1242f; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Changing a list name is different</h2>
<p style="line-height:1.8;">If the function assigns a new list to its parameter, it only changes the local name. The caller’s variable still refers to the original list.</p>

```python
def replace_list(items):
    items = [9, 9]

numbers = [1, 2]
replace_list(numbers)
print(numbers)
```

<p style="line-height:1.8;">The output is still <code>[1, 2]</code>. <code>items = [9, 9]</code> points the local name to a new list; it does not replace the caller’s list.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;"><code>append()</code>는 기존 객체를 수정하고, 대입문은 지역 이름이 가리키는 대상을 바꿉니다.</p>

<h2 style="color:#8250df; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">If you need a separate list</h2>
<p style="line-height:1.8;">Make a copy before changing the list. For a simple list, <code>items.copy()</code> creates a separate list.</p>

```python
def add_to_copy(items):
    copied_items = items.copy()
    copied_items.append(3)
    return copied_items

numbers = [1, 2]
new_numbers = add_to_copy(numbers)
print(numbers)
print(new_numbers)
```

<p style="line-height:1.8;">Now <code>numbers</code> stays <code>[1, 2]</code>, and <code>new_numbers</code> is <code>[1, 2, 3]</code>.</p>
<p style="line-height:1.8;">This is a shallow copy. If a list contains other lists, those inner lists are still shared.</p>

<h2 style="color:#1a7f37; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">A function changes a shared list</h2>
<p style="line-height:1.8;">The parameter <code>d</code> and the outside name <code>c</code> refer to the same list. Calling <code>d.append(4)</code> changes that list, so the change is visible through <code>c</code>.</p>

```python
def object_change1(d):
    d.append(4)

c = [1, 2, 3]
print(c)
object_change1(c)
print(c)
```

<p style="line-height:1.8;">The output is <code>[1, 2, 3]</code>, then <code>[1, 2, 3, 4]</code>. The function edits the list object itself.</p>
<img src="images/reference-mutation-rebinding.svg" alt="The outside name c and function parameter d point to the same list. After append, that shared list contains 4." width="700">
<p style="line-height:1.8;">When the function returns, the local name <code>d</code> disappears. The list stays because <code>c</code> still refers to it.</p>

<h2 style="color:#0550ae; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">A function assigns a new list</h2>
<p style="line-height:1.8;">Now the function assigns a new list to its parameter. This changes the local name <code>f</code>; it does not change the caller’s name <code>e</code>.</p>

```python
def object_change2(f):
    f = [4, 5, 6]

e = [1, 2, 3]
print(e)
object_change2(e)
print(e)
```

<p style="line-height:1.8;">The output is <code>[1, 2, 3]</code> both times. At the start, <code>e</code> and <code>f</code> refer to the same list. The assignment <code>f = [4, 5, 6]</code> makes <code>f</code> refer to a new list. <code>e</code> still refers to the original list.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">기존 리스트를 수정하는 것과 매개변수 이름이 새 리스트를 가리키게 하는 것은 다릅니다.</p>
<p style="line-height:1.8;">After the function ends, <code>f</code> disappears. If no name refers to the new list, Python can reclaim it.</p>
<img src="images/reference-mutation-rebinding.svg" alt="Changing the shared list affects the caller, but assigning a new list to the parameter leaves the caller pointing at the original list." width="700">

<h2 style="color:#d1242f; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Remember the difference</h2>
<ul style="line-height:1.9;">
<li><code>items.append(4)</code> changes the shared list object. The caller sees the change.</li>
<li><code>items = [4, 5, 6]</code> changes only the local parameter name. The caller keeps the original list.</li>
<li>When a function returns, its local names disappear. Objects can remain if other names still refer to them.</li>
</ul>
<p style="line-height:1.8;">The key question is: <strong>Did the function change the object, or did it point its local name to a different object?</strong></p>

<h2 style="color:#d1242f; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Easy rule</h2>
<ul style="line-height:1.9;">
<li>Assigning a new value to a parameter changes its local name.</li>
<li>Changing a shared mutable object, such as with <code>append()</code>, is visible to the caller.</li>
<li>Use <code>.copy()</code> when you need a separate list.</li>
</ul>
<p style="line-height:1.8;"><strong>Remember:</strong> Python passes object references. “Value copy” and “reference copy” are simple labels for the different results we see.</p>

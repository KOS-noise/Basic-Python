<h1 style="color:#0969da; font-size:2.4em; font-weight:700; letter-spacing:-0.02em; margin-bottom:0.35em;">Modules</h1>
<p style="line-height:1.8;">A module is a file that contains reusable Python code, such as functions and variables. We can import a module and use its contents.</p>

<h2 style="color:#8250df; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Standard and external modules</h2>
<p style="line-height:1.8;">Python comes with many modules. Together, they are called the <strong>standard library</strong>. The <code>math</code> module is one example.</p>
<p style="line-height:1.8;">Other modules are made by people outside the Python project. We can install and use them when needed.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">표준 라이브러리는 파이썬과 함께 제공되는 모듈 모음입니다. 외부 모듈은 필요할 때 따로 설치할 수 있습니다.</p>

<h2 style="color:#1a7f37; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Import a module</h2>
<p style="line-height:1.8;">Use <code>import</code> to make a module available. Write the import near the top of your file.</p>
<pre><code>import module_name</code></pre>
<p style="line-height:1.8;">After importing, use the module name and a dot to access its functions or values.</p>

```python
import math

math.sqrt(25)
```

<p style="line-height:1.8;">Modules let us reuse code instead of writing every function ourselves. They also group related tools together.</p>
<p style="line-height:1.8;">For example, <code>math.sqrt()</code> means “use the <code>sqrt()</code> function from the <code>math</code> module.”</p>

<h2 style="color:#d1242f; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Key points</h2>
<ul style="line-height:1.9;">
<li>A module groups reusable Python code.</li>
<li>The standard library comes with Python.</li>
<li>Use <code>import module_name</code> to import a module.</li>
<li>Use <code>module_name.tool</code> to access something inside it.</li>
</ul>

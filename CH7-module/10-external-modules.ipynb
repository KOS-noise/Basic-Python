<h2 style="color:#1a7f37; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Fetch weather data with Beautiful Soup</h2>
<p style="line-height:1.8;">Beautiful Soup extracts data from HTML and XML. This example reads the Korea Meteorological Administration’s RSS feed and prints the city, forecast, and minimum and maximum temperatures.</p>

[Beautiful Soup Documentation](https://beautiful-soup-4.readthedocs.io/en/latest/)

I studied Beautiful Soup in a university course for my major and used it in a project.

If you'd like to learn more, check out this repository.

[https://github.com/KOS-noise/Python_Webscraping_Analysis](https://github.com/KOS-noise/Python_Webscraping_Analysis)

The Korea Meteorological Administration discontinued its mid-term forecast RSS service on March 31, 2025. The textbook URL may no longer return the forecast XML. For live forecasts, see the [KMA API Hub](https://apihub.kma.go.kr). [KMA RSS service notice](https://www.weather.go.kr/w/pop/rss-guide.do)

```python
# 모듈을 읽어 들입니다.
from urllib import request
from bs4 import BeautifulSoup

# urlopen() 함수로 기상청의 전국 날씨를 읽습니다.
target = request.urlopen("http://www.kma.go.kr/weather/forecast/mid-term-rss3.jsp?stnId=108")

# BeautifulSoup을 사용해 웹 페이지를 분석합니다.
soup = BeautifulSoup(target, "html.parser")

# location 태그를 찾습니다.
for location in soup.select("location"):
    # 내부의 city, wf, tmn, tmx 태그를 찾아 출력합니다
    print("도시:", location.select_one("city").string)
    print("날씨:", location.select_one("wf").string)
    print("최저기온:", location.select_one("tmn").string)
    print("최고기온:", location.select_one("tmx").string)
    print()
```

Run this cell to practice parsing the textbook's sample data. It uses embedded XML and needs no network connection.

```python
from bs4 import BeautifulSoup

target = """<rss version="2.0">
  <channel>
    <item>
      <description>
        <body>
          <location><city>서울</city><data><wf>구름많음</wf><tmn>13</tmn><tmx>20</tmx></data></location>
          <location><city>인천</city><data><wf>구름많음</wf><tmn>13</tmn><tmx>19</tmx></data></location>
          <location><city>수원</city><data><wf>구름많음</wf><tmn>12</tmn><tmx>20</tmx></data></location>
        </body>
      </description>
    </item>
  </channel>
</rss>"""

# BeautifulSoup을 사용해 웹 페이지를 분석합니다.
soup = BeautifulSoup(target, "html.parser")

# location 태그를 찾습니다.
for location in soup.select("location"):
    # 내부의 city, wf, tmn, tmx 태그를 찾아 출력합니다
    print("도시:", location.select_one("city").string)
    print("날씨:", location.select_one("wf").string)
    print("최저기온:", location.select_one("tmn").string)
    print("최고기온:", location.select_one("tmx").string)
    print()
```

<h2 style="color:#0550ae; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Select tags</h2>
<p style="line-height:1.8;"><code>select()</code> finds all matching tags. <code>select_one()</code> finds the first match, and <code>.string</code> reads its text in this example. The RSS feed has one <code>location</code> tag per region. Within each location, <code>city</code> identifies the city, and the first <code>wf</code>, <code>tmn</code>, and <code>tmx</code> provide the forecast, low, and high.</p>

<h2 style="color:#1a7f37; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Create the Flask app</h2>
<p style="line-height:1.8;"><code>app = Flask(__name__)</code> creates the Flask application. <code>__name__</code> tells Flask which Python file the app is in. Flask uses this to find files and resources related to the app.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">쉽게 말해, Flask 앱을 만들고 이 앱이 어느 파일에 있는지 알려주는 코드예요.</p>

<p style="line-height:1.8;"><code>__name__</code> is a special variable that Python sets automatically. When Flask imports <code>flask_weather.py</code>, its value is <code>"flask_weather"</code>. If you run the file directly with <code>python flask_weather.py</code>, its value is <code>"__main__"</code>.</p>
<p style="line-height:1.8;">So we do not type the file name into <code>Flask()</code>. Python supplies the right module name for us. This also means we can rename the file without changing this line.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;"><code>__name__</code>은 파이썬이 자동으로 정해주는 값이에요. Flask가 파일을 불러오면 모듈 이름이 되고, 파일을 직접 실행하면 <code>"__main__"</code>이 돼요.</p>
<h2 style="color:#1a7f37; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Connect a URL to a function</h2>
<p style="line-height:1.8;"><code>@app.route("/")</code> connects the home URL, <code>/</code>, to the function below it. When a browser visits <code>http://127.0.0.1:5000/</code>, Flask calls <code>hello()</code> and sends its returned HTML to the browser.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;"><code>@app.route("/")</code>는 첫 페이지 주소를 바로 아래 함수에 연결해요. 브라우저가 그 주소에 접속하면 Flask가 함수를 실행해요.</p>

<h2 style="color:#8250df; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Show weather data in the browser</h2>
<p style="line-height:1.8;">A route connects a URL to a function. The function can return HTML text, and the browser displays that HTML as a page.</p>
<p style="line-height:1.8;">The next example reads the sample XML file with Beautiful Soup. It builds a small HTML section for each city and returns the result to the browser.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">라우트는 주소와 함수를 연결해요. 함수가 돌려준 HTML을 브라우저가 화면에 보여줘요.</p>

```python
%%writefile python-file/flask_weather.py
from pathlib import Path
from flask import Flask
from bs4 import BeautifulSoup

app = Flask(__name__)

# Load the sample XML file in this folder.
xml_path = Path(__file__).parent / "weather_sample.xml"

@app.route("/")
def weather():
    target = xml_path.read_text(encoding="utf-8")
    soup = BeautifulSoup(target, "html.parser")

    output = "<h1>Weather Forecast</h1>"

    for location in soup.select("location"):
        city = location.select_one("city").get_text(strip=True)
        weather = location.select_one("wf").get_text(strip=True)
        low = location.select_one("tmn").get_text(strip=True)
        high = location.select_one("tmx").get_text(strip=True)

        output += f"<h2>{city}</h2>"
        output += f"Weather: {weather}<br/>"
        output += f"Low / High: {low}°C / {high}°C"
        output += "<hr/>"

    return output
```

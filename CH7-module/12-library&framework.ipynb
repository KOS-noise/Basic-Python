<h1 style="color:#0969da; font-size:2.4em; font-weight:700; letter-spacing:-0.02em; margin-bottom:0.35em;">Libraries and Frameworks</h1>
<p style="line-height:1.8;">Libraries and frameworks both help us build programs with code written by others. The main difference is who controls the program flow.</p>

<h2 style="color:#8250df; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Library: we call its functions</h2>
<p style="line-height:1.8;">A library gives us useful functions. We choose when to call them and what to do with their results.</p>
<p style="line-height:1.8;">In this example, our code directly calls <code>sin()</code>, <code>cos()</code>, and other functions from the <code>math</code> library.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">라이브러리는 필요한 기능을 우리가 직접 골라서 호출해요.</p>

```python
# library
from math import sin, cos, tan, floor, ceil

print(sin(1))
print(cos(1))
print(tan(1))
print()

print(floor(2.5))
print(ceil(2.5))
```

<h2 style="color:#1a7f37; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Framework: it calls our code</h2>
<p style="line-height:1.8;">A framework provides a structure for an application. We write functions in that structure. The framework calls our functions when the right event happens.</p>
<p style="line-height:1.8;">In Flask, we write a function and connect it to a URL with <code>@app.route("/")</code>. When a browser requests that URL, Flask calls the function.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">프레임워크에서는 우리가 만든 함수를 프레임워크가 필요한 순간에 호출해요.</p>
<h3 style="color:#bf3989; font-size:1.3em; font-weight:700; margin-bottom:0.3em;">Inversion of Control (IoC)</h3>
<p style="line-height:1.8;">This change in control is called <strong>Inversion of Control</strong>. With a library, our code calls the library. With a framework, the framework calls our code.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">제어의 역전(IoC)은 프로그램의 흐름을 주도하는 쪽이 바뀌는 것을 뜻해요.</p>

```python
# framework

from flask import Flask
app = Flask(__name__)  # 현재 파일의 이름을 Flask 모듈에 전달

@app.route("/")
def hello():
    return"<h1>Hello World!</h1>"
```

<h2 style="color:#0550ae; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">What happens in the Flask example?</h2>
<ol style="line-height:1.8;">
<li><code>Flask(__name__)</code> creates the application.</li>
<li><code>@app.route("/")</code> connects the home URL to <code>hello()</code>.</li>
<li>When a browser visits the home URL, Flask calls <code>hello()</code>.</li>
<li>The function returns HTML, and Flask sends it to the browser.</li>
</ol>
<p style="line-height:1.8;">We do not call <code>hello()</code> ourselves here. Flask calls it when a request arrives. That is the framework taking control of the flow.</p>

<h2 style="color:#bf3989; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Quick summary</h2>
<ul style="line-height:1.8;">
<li><strong>Library:</strong> we call its functions when we need them. <code>math.sin(1)</code> is an example.</li>
<li><strong>Framework:</strong> we write code in its structure, and it calls our code at the right time. A Flask route is an example.</li>
<li><strong>IoC:</strong> the framework controls when our function runs.</li>
</ul>

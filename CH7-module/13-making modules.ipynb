<h1 style="color:#0969da; font-size:2.4em; font-weight:700; letter-spacing:-0.02em; margin-bottom:0.35em;">Making Python Modules</h1>
<p style="line-height:1.8;">A module is a Python file that we can import into another file. This lesson shows how to make a module and how Python treats a file when it is imported or run directly.</p>
<p style="line-height:1.8;">다른 파일에서 불러와 사용할 수 있는 파이썬 파일을 모듈이라고 합니다. 모듈을 만드는 방법과 파일을 직접 실행하거나 불러올 때의 차이를 살펴봅니다.</p>

<h2 style="color:#8250df; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Run separate Python files from a notebook</h2>

<p style="line-height:1.8;">주피터에서는 셀을 별도의 파이썬 파일로 자동 인식하지 않아요. <code>%%writefile</code>로 파일을 만들고, <code>%run</code>으로 실행하면 돼요.</p>

<h2 style="color:#1a7f37; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">1. Make a module</h2>
<p style="line-height:1.8;">This cell creates <code>test_module.py</code>. It contains a constant and functions for a circle.</p>
<p style="line-height:1.8;">아래 셀을 실행하면 코드가 <code>test_module.py</code> 파일로 저장됩니다.</p>

```python
%%writefile python-file/module_practice/test_module.py
# test_module.py

PI = 3.141592

def number_input():
    output = input("숫자 입력> ")
    return float(output)

def get_circumference(radius):
    return 2 * PI * radius

def get_circle_area(radius):
    return PI * radius * radius
```

<h2 style="color:#1a7f37; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">2. Import the module</h2>
<p style="line-height:1.8;">This cell creates <code>main.py</code>. It imports <code>test_module</code> and uses its functions. The name <code>test</code> is a shorter nickname for the module.</p>

```python
%%writefile python-file/module_practice/main.py
# main.py

import test_module as test

radius = test.number_input()
print(test.get_circumference(radius))
print(test.get_circle_area(radius))
```

<p style="line-height:1.8;">Run <code>main.py</code> from the notebook. When asked for a number, type <code>10</code>.</p>
<p style="line-height:1.8;">노트북 셀을 직접 바꾸지 않고, 저장된 <code>main.py</code> 파일을 실행합니다.</p>

```python
%run python-file/module_practice/main.py
```

<h2 style="color:#8250df; font-size:1.7em; font-weight:700; margin-bottom:0.4em;"><code>__name__ == "__main__"</code></h2>
<p style="line-height:1.8;">다른 사람들이 만든 파이썬 코드를 보다 보면 <code>__name__ == "__main__"</code>이라는 코드를 많이 볼 수 있습니다. 많은 파이썬 개발자들이 이유도 모르고 그냥 사용하는 경우가 많은데요, 이 의미가 무엇인지 자세하게 살펴보겠습니다.</p>
<p style="line-height:1.8;">You may often see <code>__name__ == "__main__"</code> in Python code. Let’s see what it means and why it is useful.</p>

<h2 style="color:#1a7f37; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">The <code>__name__</code> variable</h2>
<p style="line-height:1.8;">파이썬 코드 내부에서는 <code>__name__</code>이라는 변수를 사용할 수 있습니다. <code>__name__</code>이라는 변수에 어떤 값이 들어있는지 확인해보겠습니다.</p>
<p style="line-height:1.8;">Python creates the special variable <code>__name__</code> for each file. In a notebook, its value is usually <code>"__main__"</code>, because the notebook is running the code as the main program.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">노트북 셀에서 확인하면 보통 <code>"__main__"</code>이 나와요. 노트북이 현재 실행의 시작점이기 때문이에요.</p>

```python
__name__
```

<h2 style="color:#bf3989; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Entry point</h2>
<p style="line-height:1.8;">프로그래밍 언어에서는 프로그램의 진입점을 entry point 또는 main 이라고 부릅니다. 그리고 이러한 엔트리 포인트 또는 메인 내부에서의 <code>__name__</code>은 <code>"__main__"</code>입니다.</p>
<p style="line-height:1.8;">The <strong>entry point</strong> is where a program starts. Python sets <code>__name__</code> to <code>"__main__"</code> in the file that is run as the entry point.</p>

<h2 style="color:#1a7f37; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Print the module name</h2>

<h2 style="color:#8250df; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Why use a main guard?</h2>
<p style="line-height:1.8;">__name__ 활용하기</p>
<p style="line-height:1.8;">엔트리 포인트 파일 내부에서는 <code>__name__</code>이 <code>"__main__"</code>이라는 값을 갖습니다. 이를 활용하면 현재 파일이 모듈로 불러와졌는지, 엔트리 포인트로 실행되었는지 확인할 수 있습니다.</p>
<p style="line-height:1.8;">Use this check to tell whether a file is run directly or imported. Put test code inside <code>if __name__ == "__main__":</code> so the tests run only when you start that file directly.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">직접 실행할 때만 필요한 코드를 조건문 안에 넣어요. 다른 파일이 import하면 그 코드는 실행되지 않아요.</p>

<h2 style="color:#1a7f37; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">The problem: test code runs on import</h2>
<p style="line-height:1.8;">The next module prints test results as soon as it is imported. This can create unwanted output.</p>
<p style="line-height:1.8;">현재 <code>test_module_with_output.py</code> 파일에는 동작을 보여주기 위한 테스트 코드가 있습니다. 이 모듈을 불러오면 테스트 출력까지 함께 나옵니다.</p>

```python
%%writefile python-file/module_practice/test_module_with_output.py
# test_module_with_output.py

PI = 3.141592

def get_circumference(radius):
    return 2 * PI * radius

def get_circle_area(radius):
    return PI * radius * radius

# Example test code
print("get_circumference(10):", get_circumference(10))
print("get_circle_area(10):", get_circle_area(10))
```

```python
%%writefile python-file/module_practice/import_with_output.py
import test_module_with_output

print("The main file is running")
```

<p style="line-height:1.8;">Run the main file. Notice that the module’s test output appears during the import.</p>

```python
%run python-file/module_practice/import_with_output.py
```

<h2 style="color:#bf3989; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">The solution: add a main guard</h2>
<p style="line-height:1.8;">그런데 <code>test_module_with_output.py</code>에는 설명을 위해 추가한 테스트 코드가 있습니다. 모듈로 사용하고 있는데 내부에서 출력이 발생하니 문제가 됩니다. 이때 현재 파일이 엔트리 포인트인지 구분하는 코드를 활용합니다.</p>
<p style="line-height:1.8;">Add the test code under <code>if __name__ == "__main__":</code>. It runs when the module file starts directly, but not when another file imports it.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">조건문으로 <code>__name__</code>이 <code>"__main__"</code>인지 확인하면 돼요.</p>

```python
%%writefile python-file/module_practice/guarded_module.py
# guarded_module.py

PI = 3.141592

def get_circumference(radius):
    return 2 * PI * radius

def get_circle_area(radius):
    return PI * radius * radius

# 활용 예 (여기를 수정)
if __name__ == "__main__":
    print("get_circumference(10):", get_circumference(10))
    print("get_circle_area(10):", get_circle_area(10))
```

```python
%%writefile python-file/module_practice/import_guarded.py
import guarded_module

print("Area from the main file:", guarded_module.get_circle_area(10))
```

<p style="line-height:1.8;">Run the importing file. The guarded test prints do not appear. The main file can still use the module’s functions.</p>
<p style="line-height:1.8;">이렇게 하면 다른 파일에서 모듈을 불러와도 테스트 출력이 발생하지 않습니다.</p>

```python
%run python-file/module_practice/import_guarded.py
```

<p style="line-height:1.8;">Now run the module itself. Because it is the entry point, its test code runs.</p>

```python
%run python-file/module_practice/guarded_module.py
```

<li>A module is a <code>.py</code> file that can be imported.</li>
<li><code>__name__</code> is <code>"__main__"</code> for the file run as the entry point. An imported file gets its module name.</li>
<li><code>if __name__ == "__main__":</code> keeps test code from running during import.</li>

</ul>

<h1 style="color:#0969da; font-size:2.4em; font-weight:700; letter-spacing:-0.02em; margin-bottom:0.35em;">Making and Using Modules</h1>
<p style="line-height:1.8;">A Python module is a <code>.py</code> file. We can put functions in one file and use them from another file.</p>
<p style="line-height:1.8;">A notebook cell is not automatically a separate <code>.py</code> file. In this lesson, IPython’s <code>%%writefile</code> saves a cell as a file, and <code>%run</code> runs that file.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">노트북 셀로도 파일을 만들고 실행할 수 있어요. 파일에 쓸 코드를 바꾸지 않고 셀 맨 위에 저장 명령만 붙이면 돼요.</p>

<h2 style="color:#8250df; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">How to use the notebook</h2>
<ol style="line-height:1.8;">
<li>Run a <code>%%writefile</code> cell to save its Python code as a <code>.py</code> file.</li>
<li>Run the <code>%run</code> cell to execute that file.</li>
<li>Run these cells from the <code>CH7-module</code> folder. The examples save files in its <code>python-file</code> folder.</li>
</ol>
<p style="line-height:1.8;">The first line, such as <code>%%writefile python-file/circle_module.py</code>, is a notebook command. The lines below it are saved as the Python file.</p>

<h2 style="color:#1a7f37; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">1. Make a module</h2>
<p style="line-height:1.8;">This cell creates <code>circle_module.py</code>. It stores a constant and functions for a circle.</p>

```python
%%writefile python-file/circle_module.py
PI = 3.141592

def number_input():
    output = input("Radius > ")
    return float(output)

def get_circumference(radius):
    return 2 * PI * radius

def get_circle_area(radius):
    return PI * radius * radius
```

<h2 style="color:#1a7f37; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">2. Use the module from another file</h2>
<p style="line-height:1.8;">This cell creates <code>circle_main.py</code>. The line <code>import circle_module as circle</code> loads the module and gives it the shorter name <code>circle</code>.</p>

```python
%%writefile python-file/circle_main.py
import circle_module as circle

radius = circle.number_input()
print(circle.get_circumference(radius))
print(circle.get_circle_area(radius))
```

<p style="line-height:1.8;">Now run the main file. When the prompt appears, type a radius, such as <code>10</code>.</p>

```python
%run python-file/circle_main.py
```

<h2 style="color:#8250df; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">What does <code>__name__</code> mean?</h2>
<p style="line-height:1.8;">Python sets <code>__name__</code> automatically. If a file is run as the main program, its value is <code>"__main__"</code>. If another file imports it, its value is the module name.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">직접 실행한 파일은 <code>"__main__"</code>, import한 파일은 보통 모듈 이름을 가져요.</p>

<h2 style="color:#1a7f37; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">See the two values</h2>
<p style="line-height:1.8;">First, make a module that prints its own <code>__name__</code>.</p>

```python
%%writefile python-file/name_module.py
print("Module name:", __name__)
```

<p style="line-height:1.8;">Next, make a main file that imports the module and prints its own value.</p>

```python
%%writefile python-file/name_main.py
import name_module

print("Main program name:", __name__)
```

<p style="line-height:1.8;">Run the main file. The imported file prints <code>name_module</code>. The file run by <code>%run</code> prints <code>__main__</code>.</p>

```python
%run python-file/name_main.py
```

<h2 style="color:#bf3989; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Protect test code with a main guard</h2>
<p style="line-height:1.8;">Sometimes a module has test code at the bottom. Without a check, that code runs when the module is imported. Put it inside <code>if __name__ == "__main__":</code> to run it only when the file is started directly.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">이 조건문 안의 코드는 파일을 직접 실행할 때만 실행돼요. 다른 파일이 import하면 건너뜁니다.</p>

```python
%%writefile python-file/guarded_module.py
PI = 3.141592

def get_circumference(radius):
    return 2 * PI * radius

def get_circle_area(radius):
    return PI * radius * radius


# 이 조건문을 주목
if __name__ == "__main__":
    print("Test circumference:", get_circumference(10))
    print("Test area:", get_circle_area(10))
```

<p style="line-height:1.8;">This main file imports the module and uses its functions. The module’s test prints do not run during the import.</p>

```python
%%writefile python-file/guarded_main.py
import guarded_module

print("Area from the main file:", guarded_module.get_circle_area(10))
```

<p style="line-height:1.8;">Run the main file to see that only its result is printed:</p>

```python
%run python-file/guarded_main.py
```

<p style="line-height:1.8;">Run the module itself to see its test code:</p>

```python
%run python-file/guarded_module.py
```

<h2 style="color:#0550ae; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Quick summary</h2>
<ul style="line-height:1.8;">
<li>A module is a Python <code>.py</code> file.</li>
<li><code>%%writefile</code> saves a notebook cell as a file. Its code is not changed by the save command.</li>
<li><code>%run</code> runs a Python file from a notebook.</li>
<li><code>if __name__ == "__main__":</code> keeps test code from running when the module is imported.</li>
</ul>

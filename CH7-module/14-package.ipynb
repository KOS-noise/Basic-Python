<h1 style="color:#0969da; font-size:2.4em; font-weight:700; letter-spacing:-0.02em; margin-bottom:0.35em;">Python Packages</h1>
<p style="line-height:1.8;">A package is a folder that groups related Python modules. A module is a <code>.py</code> file. A package helps organize modules and lets us import them by name.</p>
<p style="line-height:1.8;">패키지는 관련된 모듈을 모아 둔 폴더예요. 모듈이 파일 하나라면, 패키지는 모듈을 담는 폴더라고 생각하면 돼요.</p>

<h2 style="color:#8250df; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Our package layout</h2>
<p style="line-height:1.8;">We will make two modules inside <code>test_package</code>. The <code>main.py</code> file is outside the package and uses those modules.</p>
<pre><code>module_package/
├── main.py
├── main_star.py
└── test_package/
    ├── __init__.py
    ├── module_a.py
    └── module_b.py</code></pre>
<p style="line-height:1.8;">Run these cells with the notebook’s current folder set to <code>CH7-module</code>. The files will be created inside <code>python-file/module_package</code>.</p>
<p style="line-height:1.8;">The dotted name <code>test_package.module_a</code> means: open the <code>test_package</code> package, then its <code>module_a</code> module.</p>

<h2 style="color:#1a7f37; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Prepare the folders</h2>
<p style="line-height:1.8;">Run this cell once. It creates the folders used by the examples.</p>

```python
from pathlib import Path

package_dir = Path("python-file/module_package/test_package")
package_dir.mkdir(parents=True, exist_ok=True)
```

<h2 style="color:#1a7f37; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Create two modules</h2>
<p style="line-height:1.8;">Each module defines one variable. The notebook command on the first line saves the remaining code into a Python file.</p>

```python
%%writefile python-file/module_package/test_package/module_a.py
# ./test_package/module_a.py
variable_a = "a 모듈의 변수"
```

```python
%%writefile python-file/module_package/test_package/module_b.py
# ./test_package/module_b.py
variable_b = "b 모듈의 변수"
```

<h2 style="color:#8250df; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Make <code>__init__.py</code></h2>
<p style="line-height:1.8;">A regular package uses <code>__init__.py</code> for package setup. Python runs this file when the package is imported. It can be empty, or it can set up the package.</p>
<p style="line-height:1.8;">여기서는 <code>__init__.py</code>에 패키지 초기화 메시지와 <code>__all__</code> 목록을 넣습니다.</p>

```python
%%writefile python-file/module_package/test_package/__init__.py
# Modules to load with: from test_package import *
__all__ = ["module_a", "module_b"]

print("test_package 패키지를 읽어 들였습니다.")
```

<h2 style="color:#1a7f37; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Import modules by their full names</h2>
<p style="line-height:1.8;">This main file imports each module with its full package name. It gives each imported module a short alias.</p>

```python
%%writefile python-file/module_package/main.py
# Import modules from the package.
import test_package.module_a as a
import test_package.module_b as b

# Print values from the modules.
print(a.variable_a)
print(b.variable_b)
```

<p style="line-height:1.8;">Run the main file. The package’s <code>__init__.py</code> runs first. Then <code>main.py</code> imports the two modules.</p>

```python
%run python-file/module_package/main.py
```

<h3 style="color:#0550ae; font-size:1.25em; font-weight:700;">Example result</h3>
<pre><code>test_package 패키지를 읽어 들였습니다.
a 모듈의 변수
b 모듈의 변수</code></pre>
<p style="line-height:1.8;">The first line comes from <code>__init__.py</code>. The next two lines come from <code>module_a.py</code> and <code>module_b.py</code>.</p>

<h2 style="color:#8250df; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">What does <code>__all__</code> do?</h2>
<p style="line-height:1.8;">The list <code>__all__</code> names the modules that should be imported by <code>from test_package import *</code>. Here, it lists <code>module_a</code> and <code>module_b</code>.</p>
<p style="line-height:1.8;"><code>from ... import *</code> is useful for this lesson, but it can make it hard to see where a name came from. In everyday code, importing the modules you need by name is usually clearer.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;"><code>__all__</code>은 별표로 모두 불러올 때 포함할 모듈을 정해요. 실제 코드에서는 필요한 모듈만 직접 불러오는 편이 읽기 쉬워요.</p>

```python
%%writefile python-file/module_package/main_star.py
# Import the modules listed in test_package/__init__.py.
from test_package import *

print(module_a.variable_a)
print(module_b.variable_b)
```

<p style="line-height:1.8;">Run this file to see how <code>__all__</code> controls the star import.</p>

```python
%run python-file/module_package/main_star.py
```

<h2 style="color:#bf3989; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">A note about <code>__init__.py</code></h2>
<p style="line-height:1.8;">In Python versions before 3.3, a package folder needed an <code>__init__.py</code> file. From Python 3.3 onward, Python also supports <strong>namespace packages</strong>, which can work without that file.</p>
<p style="line-height:1.8;">However, this does not mean every folder is automatically a package. Python must be able to find the folder on its import path. For this beginner example, we include <code>__init__.py</code> because it runs package setup code and defines <code>__all__</code>.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">Python 3.3 이전에는 일반적으로 <code>__init__.py</code>가 있어야 패키지로 인식됐어요. Python 3.3부터는 import 경로에서 찾을 수 있는 폴더를 namespace package로 사용할 수도 있어요.</p>
<p style="line-height:1.8;">The Python package guide explains that <code>__all__</code> controls which names are imported by <code>from package import *</code>. It also recommends importing specific modules in most code.</p>

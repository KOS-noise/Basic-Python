<h1 style="color:#0969da; font-size:2.4em; font-weight:700; letter-spacing:-0.02em; margin-bottom:0.35em;">The <code>math</code> module</h1>
<p style="line-height:1.8;">The <code>math</code> module has common math functions and constants. It is part of Python’s standard library, so no extra installation is needed.</p>

```python
import math
```

<h2 style="color:#8250df; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Explore the module</h2>
<p style="line-height:1.8;">After <code>import math</code>, type <code>math.</code> in an editor. Autocomplete can show available functions and constants.</p>
<p style="line-height:1.8;">You can also read the official Python documentation when you need details about a function.</p>

<h2 style="color:#1a7f37; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Trigonometric functions</h2>
<p style="line-height:1.8;">The module includes <code>sin()</code>, <code>cos()</code>, and <code>tan()</code>. These functions take an angle in <strong>radians</strong>, not degrees.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">삼각 함수에 각도를 넣을 때는 도(degree)가 아니라 라디안(radian)을 사용합니다.</p>

```python
print(math.sin(1))
print(math.cos(1))
print(math.tan(1))
```

<p style="line-height:1.8;">To use degrees, convert them with <code>math.radians()</code>. For example, 30 degrees is <code>π / 6</code> radians.</p>

```python
math.sin(math.radians(30))
```

<h2 style="color:#0550ae; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Floor and ceiling</h2>
<p style="line-height:1.8;"><code>math.floor(x)</code> rounds down to the greatest integer less than or equal to <code>x</code>.</p>
<p style="line-height:1.8;"><code>math.ceil(x)</code> rounds up to the smallest integer greater than or equal to <code>x</code>.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">floor는 아래쪽 정수, ceil은 위쪽 정수를 구합니다. 음수에서도 수직선 기준으로 생각하세요.</p>

```python
print(math.floor(2.5))
print(math.ceil(2.5))
print(math.floor(-2.5))
print(math.ceil(-2.5))
```

<h2 style="color:#d1242f; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Logarithms</h2>
<p style="line-height:1.8;"><code>math.log(x)</code> returns the natural logarithm of <code>x</code>. Give a second argument to choose a base: <code>math.log(x, base)</code>.</p>

```python
print(math.log(math.e))
print(math.log(8, 2))
```

<h2 style="color:#8250df; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Quick reference</h2>
<ul style="line-height:1.9;">
<li><code>math.sin(x)</code>, <code>math.cos(x)</code>, <code>math.tan(x)</code>: trigonometry; <code>x</code> is in radians.<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">각도 x의 사인, 코사인, 탄젠트를 구해요. 30도의 사인은 <code>math.sin(math.radians(30))</code>으로 구해요.</p></li>
<li><code>math.floor(x)</code>: round down.<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">x 이하인 가장 큰 정수를 구해요. 예: <code>math.floor(-3.7)</code> → <code>-4</code></p></li>
<li><code>math.ceil(x)</code>: round up.<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">x 이상인 가장 작은 정수를 구해요. 예: <code>math.ceil(-3.2)</code> → <code>-3</code></p></li>
<li><code>math.log(x)</code>: natural logarithm.<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">e를 몇 제곱하면 x가 되는지 구해요. 예: <code>math.log(math.e)</code> → <code>1.0</code></p></li>
<li><code>math.log(x, base)</code>: logarithm with the given base.<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">base를 몇 제곱하면 x가 되는지 구해요. 예: <code>math.log(8, 2)</code> → <code>3.0</code></p></li>
<li><code>math.pi</code> and <code>math.e</code>: common constants.<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;"><code>math.pi</code>는 원주율(약 3.14159), <code>math.e</code>는 자연로그의 밑(약 2.71828)이에요.</p></li>
</ul>

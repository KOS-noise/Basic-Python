<h1 style="color:#0969da; font-size:2.4em; font-weight:700; letter-spacing:-0.02em; margin-bottom:0.35em;"><code>from ... import ...</code></h1>
<p style="line-height:1.8;">Use this form when you want to import only selected functions or values from a module.</p>

<p style="line-height:1.8;">The basic form is:</p>
<pre><code>from module_name import name1, name2</code></pre>
<p style="line-height:1.8;">After this import, use the selected names directly. You do not need to write the module name each time.</p>

```python
from math import sin, cos, tan, floor, ceil

print(sin(1))
print(cos(1))
print(tan(1))
print(floor(2.5))
print(ceil(2.5))
```

<h2 style="color:#8250df; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Two ways to use a function</h2>
<pre><code>import math
math.sin(1)

from math import sin
sin(1)</code></pre>
<p style="line-height:1.8;">Both forms use the same function. The first keeps its module name visible. The second is shorter.</p>

<h2 style="color:#1a7f37; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Avoid importing everything</h2>
<p style="line-height:1.8;">You can write <code>from math import *</code>, but it is usually better to import only the names you need.</p>
<p style="line-height:1.8;">Importing everything can make it unclear where a name came from. It can also create name conflicts.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;"><code>*</code>는 전부 가져온다는 뜻이지만, 이름 충돌과 출처 혼란을 막으려면 필요한 것만 가져오는 편이 좋습니다.</p>

```python
# Usually avoid this:
# from math import *
```

<h2 style="color:#d1242f; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Key points</h2>
<ul style="line-height:1.9;">
<li>Use <code>from module import name</code> to import selected names.</li>
<li>Use the imported names directly, without the module prefix.</li>
<li>Import only what you need to keep code clear.</li>
</ul>

<h1 style="color:#0969da; font-size:2.4em; font-weight:700; letter-spacing:-0.02em; margin-bottom:0.35em;"><code>round()</code></h1>
<p style="line-height:1.8;"><code>round()</code> is a built-in function. It rounds a number to the nearest integer or to a chosen number of decimal places.</p>

<h2 style="color:#8250df; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Basic use</h2>
<p style="line-height:1.8;">Use <code>round(number)</code> to round to the nearest integer. Use <code>round(number, digits)</code> to keep a chosen number of decimal places.</p>

```python
print(round(1.4))
print(round(1.6))
print(round(3.14159, 2))
```

<h2 style="color:#1a7f37; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">When the value is halfway</h2>
<p style="line-height:1.8;">When a value is exactly halfway between two integers, Python chooses the even one. This is called <strong>round half to even</strong>.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">정확히 가운데에 걸린 값은 가까운 짝수 쪽으로 반올림합니다.</p>

```python
print(round(1.5))
print(round(2.5))
print(round(3.5))
print(round(4.5))
```

<p style="line-height:1.8;">Decimal fractions are not always exact in binary floating-point. So a result may look surprising: <code>round(2.675, 2)</code> returns <code>2.67</code>.</p>
<p style="line-height:1.8;">For exact decimal calculations, such as money, use the <code>decimal</code> module.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">컴퓨터의 부동소수점 표현 때문에 일부 소수는 정확히 저장되지 않습니다. 그래서 특정 소수의 반올림 결과가 예상과 다를 수 있습니다.</p>

```python
round(2.675, 2)
```

<h2 style="color:#d1242f; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Key points</h2>
<ul style="line-height:1.9;">
<li><code>round(x)</code> rounds to the nearest integer.</li>
<li><code>round(x, n)</code> rounds to <code>n</code> decimal places.</li>
<li>Exact halfway cases go to the even choice.</li>
<li>Binary floating-point can affect decimal-looking values.</li>
</ul>

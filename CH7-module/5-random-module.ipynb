<h1 style="color:#0969da; font-size:2.4em; font-weight:700; letter-spacing:-0.02em; margin-bottom:0.35em;"><code>random</code></h1>
<p style="line-height:1.8;">The <code>random</code> module makes pseudo-random numbers and selects random items. It is part of Python’s standard library.</p>

<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">Pseudo-random 은 의사 난수, 즉 무작위처럼 보이도록 계산해서 만든 숫자라는 뜻. 예를 들어 파이썬의 random.randint(1, 6)은 주사위를 굴린 것처럼 1부터 6까지의 숫자 중 하나를 뽑는다. 이 숫자들은 시드(seed)라는 시작값을 바탕으로 만들어진다.</p>

```python
import random
```

<h2 style="color:#8250df; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Random numbers</h2>
<p style="line-height:1.8;"><code>random()</code> returns a float from <code>0.0</code> up to, but not including, <code>1.0</code>. <code>uniform(a, b)</code> returns a float between two bounds. <code>randrange()</code> picks an integer from a range.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">These results can change each time you run the code.</p>

```python
print(random.random())
print(random.uniform(10, 20))
print(random.randrange(10))       # 0 to 9
print(random.randrange(1, 10))    # 1 to 9
```

<h2 style="color:#1a7f37; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Choose and shuffle items</h2>
<p style="line-height:1.8;"><code>choice()</code> picks one item. <code>shuffle()</code> changes a list in place. <code>sample()</code> returns a new list of unique picks.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">In <code>shuffle()</code>, the original list changes. It returns <code>None</code>.</p>

```python
items = [1, 2, 3, 4, 5]

print(random.choice(items))

random.shuffle(items)
print(items)

winners = random.sample(items, k=2)
print(winners)
```

<h2 style="color:#bf3989; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Important</h2>
<p style="line-height:1.8;">These functions are useful for games, simple simulations, and practice. They are not for passwords or security tokens. Use Python’s <code>secrets</code> module for security-sensitive random values.</p>
<p style="line-height:1.8;"><strong>Remember:</strong> <code>choice</code> picks one, <code>shuffle</code> rearranges the list, and <code>sample</code> picks several without repeating an item.</p>

<h1 style="color:#0969da; font-size:2.4em; font-weight:700; letter-spacing:-0.02em; margin-bottom:0.35em;"><code>sys</code></h1>
<p style="line-height:1.8;">The <code>sys</code> module gives information about the Python interpreter and the program running in it.</p>

```python
import sys
```

<h2 style="color:#8250df; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Command-line arguments</h2>
<p style="line-height:1.8;"><code>sys.argv</code> is a list of text values passed to a Python script from the command line. The first item is the script name. The next items are the values you type after it.</p>
<p style="line-height:1.8;">For example, run <code>python greet.py Mina 20</code>. The list is similar to <code>['greet.py', 'Mina', '20']</code>. In a notebook, the values depend on how the notebook was started.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">Arguments usually arrive as strings, so convert them before doing math.</p>

```python
# Example values for a script started like: python greet.py Mina 20
example_argv = ["greet.py", "Mina", "20"]

script_name = example_argv[0]
name = example_argv[1]
age = int(example_argv[2])

print(script_name)
print(f"Hello, {name}. Next year you will be {age + 1}.")
```

<h2 style="color:#1a7f37; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Interpreter information</h2>
<p style="line-height:1.8;">The following values show the Python version and the platform running this code. Some details depend on your computer.</p>

```python
print(sys.version)
print(sys.platform)
print(sys.argv)
```

<h2 style="color:#bf3989; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Ending a program</h2>
<p style="line-height:1.8;"><code>sys.exit()</code> asks a script to stop. It raises a special exit signal, so we usually use it in a script when the program should end. Avoid running it in a notebook cell because it can interrupt the notebook session.</p>
<p style="line-height:1.8;">Use the standard library documentation when you need a function’s exact behavior.</p>

<h1 style="color:#0969da; font-size:2.4em; font-weight:700; letter-spacing:-0.02em; margin-bottom:0.35em;"><code>os</code></h1>
<p style="line-height:1.8;">The <code>os</code> module provides tools for working with the operating system, folders, and file paths.</p>
<p style="line-height:1.8;">File and folder operations depend on the computer and the current working folder. This lesson uses a temporary folder so the examples clean up after themselves.</p>

```python
import os
from pathlib import Path
from tempfile import TemporaryDirectory
```

<h2 style="color:#8250df; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Check the current folder</h2>
<p style="line-height:1.8;"><code>getcwd()</code> returns the current working directory. <code>listdir()</code> returns the names inside a folder. The exact output depends on where Python is running.</p>

```python
print("Current folder:", os.getcwd())
print("Items here:", os.listdir())
```

<h2 style="color:#1a7f37; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Create, rename, and remove files</h2>
<p style="line-height:1.8;">The example makes a temporary folder, creates a file, renames it, and removes it. When the <code>with</code> block ends, Python removes the temporary folder and its contents.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">Be careful with <code>remove()</code> and <code>rmdir()</code>: they delete files or folders.</p>

```python
with TemporaryDirectory() as temp_dir:
    folder = Path(temp_dir)
    demo_folder = folder / "hello"
    demo_folder.mkdir()

    old_file = demo_folder / "original.txt"
    new_file = demo_folder / "new.txt"

    old_file.write_text("hello", encoding="utf-8")
    os.rename(old_file, new_file)

    print("Renamed file:", new_file.name)
    print("File contents:", new_file.read_text(encoding="utf-8"))

    os.remove(new_file)
    os.rmdir(demo_folder)
    print("File and folder removed")
```

<h2 style="color:#bf3989; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Use paths carefully</h2>
<p style="line-height:1.8;">A relative path such as <code>"notes.txt"</code> starts from the current working directory. Check the path before changing or deleting anything.</p>
<p style="line-height:1.8;">For portable path handling, <code>pathlib</code> is often easier to read. The example uses it to build paths, and uses <code>os</code> to show the module functions from this lesson.</p>

<h2 style="color:#8250df; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Delete files and folders</h2>
<p style="line-height:1.8;"><code>os.remove()</code> deletes a file. <code>os.unlink()</code> does the same thing. Use <code>os.rmdir()</code> to delete an empty folder. These functions cannot restore deleted items.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">Always check the path before deleting. A wrong path can remove important data.</p>

<h2 style="color:#bf3989; font-size:1.7em; font-weight:700; margin-bottom:0.4em;"><code>os.system()</code> and safety</h2>
<p style="line-height:1.8;"><code>os.system(command)</code> runs a command through the operating system’s shell. The command can do anything your account is allowed to do, including deleting files.</p>
<p style="line-height:1.8;">Do not run commands you do not understand. Avoid putting user input into a shell command. For many common tasks, use Python functions such as <code>Path</code>, <code>os</code>, or <code>subprocess</code> with an argument list instead.</p>

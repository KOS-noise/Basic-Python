<h1 style="color:#0969da; font-size:2.4em; font-weight:700; letter-spacing:-0.02em; margin-bottom:0.35em;"><code>datetime</code></h1>
<p style="line-height:1.8;">The <code>datetime</code> module works with dates and times. It can get the current time and format it as text.</p>

```python
import datetime
```

<h2 style="color:#8250df; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Get the current date and time</h2>
<p style="line-height:1.8;">Use <code>datetime.datetime.now()</code> to get the local date and time. The value contains fields such as <code>year</code>, <code>month</code>, <code>day</code>, <code>hour</code>, <code>minute</code>, and <code>second</code>.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">The output changes with the current time and your computer’s time zone.</p>

```python
now = datetime.datetime.now()

print("Year:", now.year)
print("Month:", now.month)
print("Day:", now.day)
print("Hour:", now.hour)
print("Minute:", now.minute)
print("Second:", now.second)
```

<h2 style="color:#1a7f37; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Format the date and time</h2>
<p style="line-height:1.8;">Use <code>strftime()</code> to turn a date and time into a string. The format codes begin with <code>%</code>.</p>
<ul style="line-height:1.8;"><li><code>%Y</code>: four-digit year</li><li><code>%m</code>: month</li><li><code>%d</code>: day</li><li><code>%H</code>: hour in 24-hour format</li><li><code>%M</code>: minute</li><li><code>%S</code>: second</li></ul>

```python
now = datetime.datetime.now()

formatted = now.strftime("%Y.%m.%d %H:%M:%S")
print(formatted)

friendly = now.strftime("%Y-%m-%d at %H:%M")
print(friendly)
```

<h2 style="color:#bf3989; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">A format string is a pattern</h2>
<p style="line-height:1.8;">In <code>"%Y.%m.%d %H:%M:%S"</code>, the codes are replaced with values from <code>now</code>. The dots, spaces, and colons stay as written.</p>
<p style="line-height:1.8;">Use <code>strftime()</code> when you want to display a date. Use <code>strptime()</code> when you need to read text and turn it into a date object.</p>

<h1 style="color:#0969da; font-size:2.4em; font-weight:700; letter-spacing:-0.02em; margin-bottom:0.35em;">The <code>time</code> and <code>urllib</code> modules</h1>
<p style="line-height:1.8;">Use <code>time</code> to work with timestamps and pause execution. Use <code>urllib.request</code> to fetch data from a URL.</p>

<h2 style="color:#8250df; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Unix time</h2>
<p style="line-height:1.8;"><code>time.time()</code> returns the current Unix timestamp in seconds since January 1, 1970, at 00:00 UTC. The result can include a fractional part.</p>

```python
import time

print(time.time())
```

<h2 style="color:#1a7f37; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Pause with <code>time.sleep()</code></h2>
<p style="line-height:1.8;"><code>time.sleep()</code> pauses execution for the requested duration. Pass the number of seconds you want to wait, such as <code>5</code> or <code>0.5</code>.</p>

```python
print("Pausing for 5 seconds...")
time.sleep(5)
print("Done.")
```

<h2 style="color:#0550ae; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Fetch a URL with <code>urllib.request</code></h2>
<p style="line-height:1.8;"><code>urllib</code> provides tools for working with URLs. URL stands for <strong>Uniform Resource Locator</strong>. It specifies where a resource can be found on a network.</p>
<p style="line-height:1.8;"><code>request.urlopen()</code> opens the URL and returns a response. <code>read(200)</code> reads up to 200 bytes from the response body. The <code>with</code> block closes the response automatically.</p>

```python
from urllib import request

with request.urlopen("https://github.com/KOS-noise", timeout=10) as response:
    preview = response.read(200)

print(preview)
```

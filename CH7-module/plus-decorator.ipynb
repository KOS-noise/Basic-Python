<h1 style="color:#0969da; font-size:2.4em; font-weight:700; letter-spacing:-0.02em; margin-bottom:0.35em;">Function Decorators</h1>
<p style="line-height:1.8;">A decorator adds behavior to a function without changing the function’s main code.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">데코레이터는 함수의 원래 코드를 크게 바꾸지 않고 기능을 덧붙여요.</p>

<h2 style="color:#8250df; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">The basic idea</h2>
<p style="line-height:1.8;">A decorator receives a function and returns a new function. The new function can run extra code before or after the original function.</p>
<p style="line-height:1.8;">Writing <code>@show_messages</code> above a function is a short form of <code>greet = show_messages(greet)</code>.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">함수를 데코레이터에 전달하고, 데코레이터가 돌려준 함수로 바꿔 연결하는 방식이에요.</p>

<h2 style="color:#1a7f37; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">A simple decorator</h2>
<p style="line-height:1.8;">Here, <code>show_messages</code> adds a message before and after <code>greet()</code>. The <code>wrapper</code> function calls the original function.</p>

```python
from functools import wraps

def show_messages(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        print("Starting the function")
        result = func(*args, **kwargs)
        print("The function is done")
        return result
    return wrapper

@show_messages
def greet(name):
    """Say hello to a person."""
    print(f"Hello, {name}!")
    return "Greeting complete"

result = greet("Mina")
print(result)
print(greet.__name__)
```

<h2 style="color:#bf3989; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Why import <code>wraps</code>?</h2>
<p style="line-height:1.8;">The decorator replaces the original function with <code>wrapper</code>. Without extra help, the function’s name and docstring can look like they belong to <code>wrapper</code>.</p>
<p style="line-height:1.8;"><code>from functools import wraps</code> gives us <code>@wraps(func)</code>. It keeps useful information from the original function, such as its name and docstring.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;"><code>wraps</code>는 겉으로 보이는 함수 이름과 설명이 원래 함수 정보로 유지되게 도와줘요.</p>
<p style="line-height:1.8;">The <code>*args</code> and <code>**kwargs</code> let the wrapper pass along different positional and keyword arguments.</p>

<h2 style="color:#0550ae; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">A decorator with its own parameter</h2>
<p style="line-height:1.8;">Sometimes we want to give a setting to the decorator. For example, <code>@repeat(3)</code> means “run this function three times.”</p>
<p style="line-height:1.8;">This needs one extra function: <code>repeat(times)</code> receives the setting, <code>decorator(func)</code> receives the function, and <code>wrapper()</code> runs it.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">데코레이터에 설정값을 전달할 때는 바깥 함수가 설정값을 받고, 안쪽 함수가 꾸밀 함수를 받아요.</p>

```python
from functools import wraps

def repeat(times):
    def decorator(func):
        @wraps(func)
        def wrapper(*args, **kwargs):
            result = None
            for _ in range(times):
                result = func(*args, **kwargs)
            return result
        return wrapper
    return decorator

@repeat(3)
def say_hello(name):
    print(f"Hello, {name}!")

say_hello("Mina")
```

<h2 style="color:#1a7f37; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Why decorators are useful</h2>
<p style="line-height:1.8;">Decorators are useful when many functions need the same extra behavior, such as logging, timing, or checking access. Put the shared behavior in one decorator instead of repeating it in every function.</p>
<p style="line-height:1.8;">This can make the main function code shorter and easier to read. Use a decorator when it makes the code clearer; a small one-time change may be simpler without one.</p>

<h2 style="color:#bf3989; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Python decorators and Java annotations</h2>
<p style="line-height:1.8;">Both use the <code>@</code> symbol, but they do different jobs. A Python decorator is code: it receives a function and can replace it with another function.</p>
<p style="line-height:1.8;">A Java annotation usually marks code with metadata. A compiler or framework may read that information and act on it. The annotation itself does not normally wrap the method like a Python decorator does.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">기호는 같아도 역할은 달라요. 파이썬 데코레이터는 함수를 감싸거나 바꾸는 코드이고, Java 어노테이션은 보통 코드에 붙이는 정보 표시예요.</p>

<h2 style="color:#8250df; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Quick summary</h2>
<ul style="line-height:1.8;">
<li><code>@decorator</code> adds behavior to a function.</li>
<li><code>wraps</code> keeps the original function’s name and docstring.</li>
<li>A decorator can take settings, such as <code>@repeat(3)</code>.</li>
<li>Decorators help avoid repeating the same extra code.</li>
</ul>

<h1 style="color:#0969da; font-size:2.4em; font-weight:700; letter-spacing:-0.02em; margin-bottom:0.35em;">The <code>itemgetter()</code> function</h1>
<p style="line-height:1.8;"><code>operator.itemgetter()</code> creates a function that retrieves an item by index or key. For example, <code>itemgetter(1)</code> creates a function that retrieves the second item of a list or tuple.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">인덱스나 키를 지정하면, 해당 값을 꺼내는 함수를 만듭니다.</p>

```python
from operator import itemgetter

student = ("Mina", 90)
get_score = itemgetter(1)
print(get_score(student))  # 90
```

<h2 style="color:#8250df; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Sort by an item</h2>
<p style="line-height:1.8;">Pass <code>itemgetter(1)</code> as the <code>key</code> to <code>sorted()</code> to sort students by score.</p>

```python
students = [student, ("Jisoo", 82), ("Alex", 95)]
print(sorted(students, key=get_score))
```

<h2 style="color:#1a7f37; font-size:1.7em; font-weight:700; margin-bottom:0.4em;">Find the cheapest and most expensive books</h2>
<p style="line-height:1.8;">In the CH5 example, <code>lambda book: book["price"]</code> retrieves the price from each book dictionary. <code>itemgetter("price")</code> creates an equivalent key function. Pass it to <code>min()</code> and <code>max()</code> to compare prices and return the corresponding book dictionaries.</p>
<p style="font-size:0.85em; opacity:0.7; margin:0.15em 0 1em 0;">CH5의 람다와 같은 역할을 합니다. 가격으로 비교하고 선택된 책의 딕셔너리를 반환합니다.</p>

```python
from operator import itemgetter

books = [
    {"title": "river boy", "price": 18000},
    {"title": "the giver", "price": 26000},
    {"title": "the secret garden", "price": 24000},
]

cheapest = min(books, key=itemgetter("price"))
most_expensive = max(books, key=itemgetter("price"))

print("Cheapest:", cheapest)
print("Most expensive:", most_expensive)
```
